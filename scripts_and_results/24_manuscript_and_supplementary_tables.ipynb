# Manuscript and supplementary tables

**Run All** with the notebook in the directory containing `random_outerfolds/`, or set
`PROJECT_DIR` in the first code cell. `datasets/` is found in that directory or its parent.
All result and dataset paths are derived from these roots. File paths and source-column
names are configuration; counts, metric values, hyperparameters and selected thresholds
are always read/calculated from the actual inputs.

The publication schemas retain Main Tables 1–7 and Supplementary Tables S1–S21. Each
table selects only its specified models, populations, metrics and folds. Extra fields in
combined CSV files never expand the publication tables. Excluded reference tables
(Main 6, S1 and S17) retain their numbers.

The supplied training outputs use `evaluation_subset` with populations such as
`classification_all`, and combine wide metric rows with long `metric,value` rows.
Both are read correctly. Parameter dictionaries in `parameters_json` and
`best_parameters_json` are resolved per selected model and row. The model-4 alias in
training outputs is matched to the custom-MTL evaluation role without borrowing another
model's statistics.

| Table | Source and calculation |
|---|---|
| Main 1–3, S13–S15 | Saved pooled estimates and paired statistics in `all_model_eval_main` |
| S4 | `19_kpuu_statistics.py`: `transfer_metrics.csv` and `paired_differences_with_pvalues.csv` |
| S5/S6 | `21_herg_statistics.py`: `metrics_with_ci.csv` and `paired_comparisons.csv` |
| Main 7 | Class totals and measured-logBB counts from the original/training/external datasets |
| S7 | Original training feature matrix/list, when supplied, and retained names from `run_signature.json` |
| S8 | Search-space definitions parsed from the training code specified by `TRAINING_CODE_FILE` |
| S9 | Saved selected model-4 parameters in `all_models_best_parameters.csv` |
| S10 | Model-4 `measured_regression` rows in `all_models_fold_metrics.csv` |
| S11 | Model-4 `classification_all` rows in the same file |
| S12 | Saved `tuned_threshold` values in the five threshold-model JSON files; parameter CSV fallback |

S10/S11 show the unweighted mean ± sample SD of the saved, unrounded fold values.
Saved folds 0–4 are displayed as folds 1–5; rows are never reordered to reproduce old paper
numbers. S13/S14 follow their classification-only/all-compound captions, respectively.
Performance metrics, p-values, correction families and CIs are not recomputed. Models are
not refitted and thresholds are not selected again.

  The training dataset is used only for the **Data set used** row of Main 7.

tables from exporting. No numbers are copied from the proof to fill missing inputs. In
particular, the unrelated external descriptor matrix is not used as the original training
matrix for S7.

## 2. Read saved estimates and comparisons

The original 09/11 loaders and numerical-consistency checks are retained. Native
`kpuu` and `herg` readers accept the attached statistics scripts' outputs directly;
they do not need to be converted to `09` or merged into another CSV. Reverse-order
saved comparisons have their difference and CI signs reversed; their two-sided p-values
remain unchanged. Exact model pairs and subsets must match. Missing comparisons are not
borrowed from another model alias, metric, or population.

An optional `format="long"` adapter supports other tasks without guessing which column
contains p-values. Its estimate CSV has `model, subset, metric, value` and optionally `n`.
Its pairwise CSV has `model_a, model_b, subset, metric, p_adjusted`; optional columns include
`value_a, value_b, difference, ci_low, ci_high, n, p_raw, confidence, test_method, holm_family`.
Use `estimates_file`, `pairwise_files`, `estimate_column_map`, `pair_column_map`, `metric_map`,
or `subset_map` in a source entry to map actual saved fields explicitly. `p_adjusted` must
contain adjusted p-values in [0,1], never a metric difference. Bounded or unavailable
values are not invented. The correction method/family should be supplied as metadata.

For hERG, use distinct canonical metrics `average_precision` and `pr_auc_trapezoid`.
The Kp,uu template uses `spearman_r, pearson_r, kendall_tau, roc_auc, pr_auc_nonpenetrant,
balanced_accuracy, np_precision, np_recall, np_f1, specificity, mcc, cohens_kappa`.
The low-Kp,uu class is positive. Label changes do not change any saved measurement.

Wide format 11 sources also accept `estimate_column_map` to read equivalent headers from an existing metrics CSV. BBBP uses this mapping and requires no paired-comparison file.


```python
CLASS_METRICS = ["balanced_accuracy", "np_recall", "np_precision", "np_f1", "roc_auc", "pr_auc_nonpenetrant", "cohens_kappa", "mcc", "np_wrong_side_count", "wrong_side_mean_distance"]
REG_METRICS = ["mae", "mse", "pearson_r", "r2", "np_mae", "wrong_side_mean_distance"]
LOWER_IS_BETTER = {"mae", "mse", "np_mae", "np_wrong_side_count", "wrong_side_mean_distance"}
TUNED_LIGHTGBM = "lightgbm_mse_regression_threshold"
LABELS = {"balanced_accuracy": "Balanced Accuracy", "np_recall": "Recall", "np_precision": "Precision", "np_f1": "F1",
          "roc_auc": "ROC AUC", "pr_auc_nonpenetrant": "PR AUC", "cohens_kappa": "Cohen’s Kappa", "mcc": "MCC",
          "np_wrong_side_count": "NP wrong side (FN)", "wrong_side_mean_distance": "Mean distance from margin (incorrect)",
          "mae": "MAE", "mse": "MSE", "pearson_r": "PCC", "r2": "R²", "np_mae": "Non-penetrant MAE"}
METRIC_ALIASES = {"cohen_kappa": "cohens_kappa", "cohen_kappa_score": "cohens_kappa", "kappa": "cohens_kappa", "pr_auc": "pr_auc_nonpenetrant",
                  "pr_auc_np": "pr_auc_nonpenetrant", "mae_nonpenetrant": "np_mae", "fn": "np_wrong_side_count",
                  "wrong_side_mean_distance_from_margin": "wrong_side_mean_distance"}
SUBSET_ALIASES = {"classification_only_threshold": "classification_only", "all_threshold": "all_classification",
                  "measured_threshold": "measured_classification", "classification": "all_classification",
                  "regression": "measured_regression", "all": "all_classification", "all_nonoverlapping": "all_classification"}
SUBSET_LABELS = {"classification_only": "Classification-only compounds", "all_classification": "All classification-labelled compounds",
                 "measured_regression": "Molecules with experimental logBB labels", "measured_classification": "Classification on measured-logBB compounds"}
REFERENCE_CANDIDATES = ["logbb_pred_matched_retraining", "logbb_pred_shaker_retrained", "logbb_pred_matched", "logbb_pred_retrained"]
CUSTOM_MODEL_CANDIDATES = ["custom_mtl_bab", "custom_mtl_original", "custom_mtl"]
MODEL_LABELS = {"dummy_prior_classification": "1c: Classification baseline", "dummy_mean_regression": "1r: Regression baseline",
                "lightgbm_mse_regression": "3: LightGBM (MSE objective)", "custom_mtl_bab": "4: This model (custom MTL)",
                "lightgbm_mse_regression_threshold": "5: LightGBM (MSE; tuned threshold)"}
for name in CUSTOM_MODEL_CANDIDATES:
    MODEL_LABELS[name] = "4: This model (custom MTL)"
for name in REFERENCE_CANDIDATES:
    MODEL_LABELS[name] = "2: Retrained logBB_pred"
for prefix, label in [("imagemol_resnet18_frozen", "ImageMol (frozen)"), ("unimolv1_frozen", "Uni-Mol v1 (frozen)")]:
    for task, loss in [("regression", "MSE"), ("custom_mtl", "custom MTL")]:
        MODEL_LABELS[f"{prefix}_{task}"] = f"{label}; {loss}"
        for suffix, threshold in [("native_threshold", "native threshold"), ("altered_threshold", "tuned threshold")]:
            MODEL_LABELS[f"{prefix}_{task}_{suffix}"] = f"{label}; {loss}; {threshold}"
DISTANCE_09 = "Distance is the saved mean |predicted logBB + 1| for mistakes defined at the fixed −1 cutoff, on measured-logBB compounds only. It is not based on the tuned decision rule."
DISTANCE_11 = "Distance is the saved mean |predicted logBB + 1| among mistakes defined by each model's actual decisions, on the table's classification population."

def model_label(model):
    return LABEL_OVERRIDES.get(model, MODEL_LABELS.get(model, model))

def metric_name(name):
    return METRIC_ALIASES.get(str(name), str(name))

SUBSET_ALIASES.update({
    "classification_all":"all_classification", "classification_measured":"measured_classification",
    "all_classification":"all_classification", "measured_regression":"measured_regression",
})

def subset_name(name):
    return SUBSET_ALIASES.get(str(name), str(name))

def choose_column(frame, names, default=np.nan):
    for name in names:
        if name in frame:
            return frame[name]
    return pd.Series(default, index=frame.index)

def numbers(series):
    return pd.to_numeric(series, errors="raise")

def read_optional(path):
    path = Path(path)
    return pd.read_csv(path, low_memory=False) if path.exists() else pd.DataFrame()

def location(source, key, default):
    return Path(source[key]) if key in source else Path(source["directory"]) / default

def parse_pairs(raw, source, source_path, confidence):
    if raw.empty:
        return pd.DataFrame()
    for column in ["model_a", "model_b", "metric"]:
        if column not in raw:
            raise ValueError(f"{source_path}: missing {column}.")
    out = raw[["model_a", "model_b", "metric"]].copy()
    out["dataset"] = source["dataset"]
    out["subset"] = choose_column(raw, ["task", "subset"], "all").map(subset_name) if "task" in raw else choose_column(raw, ["subset"], "all").map(subset_name)
    out["metric"] = out.metric.map(metric_name)
    out["value_a"] = numbers(choose_column(raw, ["model_a_value", "metric_a"]))
    out["value_b"] = numbers(choose_column(raw, ["model_b_value", "metric_b"]))
    out["difference"] = numbers(choose_column(raw, ["difference_a_minus_b"]))
    out["ci_low"] = numbers(choose_column(raw, ["ci_lower", "ci_low"]))
    out["ci_high"] = numbers(choose_column(raw, ["ci_upper", "ci_high"]))
    out["n"] = numbers(choose_column(raw, ["n_paired"]))
    if "permutation_p_two_sided_holm_pair_dataset" in raw:
        out["p_adjusted"] = numbers(raw.permutation_p_two_sided_holm_pair_dataset)
        out["p_raw"] = numbers(choose_column(raw, ["permutation_p_two_sided"]))
        out["test_method"], out["holm_family"] = "paired two-sided permutation test", "all evaluated metrics within each model pair and subset"
    elif "permutation_p_two_sided_holm" in raw:
        out["p_adjusted"] = numbers(raw.permutation_p_two_sided_holm)
        out["p_raw"] = numbers(choose_column(raw, ["permutation_p_two_sided"]))
        out["test_method"], out["holm_family"] = "paired two-sided permutation test", "all evaluated metrics within each model pair and task"
    elif "p_bootstrap_holm" in raw:
        out["p_adjusted"] = numbers(raw.p_bootstrap_holm)
        out["p_raw"] = numbers(choose_column(raw, ["p_bootstrap_two_sided"]))
        out["test_method"], out["holm_family"] = "paired two-sided bootstrap test", "model pairs within each subset and metric"
    else:
        raise ValueError(f"{source_path}: no recognized adjusted p-value column; raw p-values will not be substituted.")
    out["confidence"] = confidence
    out["source_file"] = str(source_path)
    out["source_record"] = np.arange(1, len(raw) + 1)  # 1-based data record, excluding the header.
    out["record_type"] = "comparison"
    return out

# Additional task metrics are accepted only when explicitly present in saved results.
EXTRA_METRICS = {'accuracy', 'mde', 'spearman_r', 'kendall_tau', 'specificity', 'average_precision',
                 'pr_auc_trapezoid', 'blocker_precision', 'blocker_recall', 'blocker_f1',
                 'nonblocker_recall', 'nonblocker_precision', 'nonblocker_f1', 'misclassification_rate',
                 'reported_pr_auc_nonpenetrant'}
LABELS.update({metric: metric.replace('_', ' ').capitalize() for metric in EXTRA_METRICS})
METRIC_ALIASES.update({'median_absolute_error': 'mde', 'median_abs_error': 'mde',
                       'spearman': 'spearman_r', 'spearman_rho': 'spearman_r',
                       'kendalltau': 'kendall_tau', 'kendall_tau_b': 'kendall_tau',
                       'matthews_corrcoef': 'mcc', 'cohen_kappa_score': 'cohens_kappa'})
LOWER_IS_BETTER.update({'mde', 'misclassification_rate'})
ALL_METRICS = set(CLASS_METRICS + REG_METRICS) | EXTRA_METRICS

def load_normalized_source(source):
    """Read explicit saved long estimates/pairwise statistics for e.g. Kp,uu or hERG.

    estimates: model, subset, metric, value, n (optional).
    pairwise: model_a, model_b, subset, metric, p_adjusted; optional
    value_a, value_b, difference, ci_low, ci_high, n, p_raw, confidence,
    test_method, holm_family. The p_adjusted column must contain actual
    adjusted p-values, never a metric difference. No correction is performed here.
    """
    directory = Path(source['directory'])
    path = location(source, 'estimates_file', 'estimates.csv')
    if not path.exists():
        if source.get('required', True):
            raise FileNotFoundError(path)
        return pd.DataFrame(), pd.DataFrame(), [f"{source['dataset']}: optional source missing ({path})."]
    data = pd.read_csv(path).rename(columns=source.get('estimate_column_map', {}))
    required = {'model', 'subset', 'metric', 'value'}
    if not required.issubset(data):
        raise ValueError(f'{path}: expected long estimate columns {sorted(required)}.')
    metric_map, subset_map = source.get('metric_map', {}), source.get('subset_map', {})
    normalize_metric = lambda s: metric_name(metric_map.get(s, s))
    normalize_subset = lambda s: subset_name(subset_map.get(s, s))
    result = data[['model', 'subset', 'metric', 'value']].copy()
    result['subset'] = result['subset'].map(normalize_subset)
    result['metric'] = result['metric'].map(normalize_metric)
    result['value'] = numbers(result['value'])
    result['n'] = numbers(choose_column(data, ['n', 'n_samples']))
    result['dataset'], result['format'] = source['dataset'], 'long'
    result['source_file'], result['metric_model'] = str(path), result['model']
    result['configured_reference'] = source.get('reference', '')
    result['distance_definition'] = source.get('distance_definition', 'Definition retained from the upstream evaluator.')
    result['record_type'] = 'estimate'
    paths = [Path(p) for p in source['pairwise_files']] if 'pairwise_files' in source else [directory/'pairwise.csv']
    pairs, notices = [], []
    for pair_path in paths:
        if not pair_path.exists():
            notices.append(f"{source['dataset']}: pairwise source missing ({pair_path}).")
            continue
        frame = pd.read_csv(pair_path).rename(columns=source.get('pair_column_map', {}))
        required = {'model_a', 'model_b', 'subset', 'metric', 'p_adjusted'}
        if not required.issubset(frame):
            raise ValueError(f'{pair_path}: missing {required-set(frame)}. Supply an explicit adjusted-p column, not differences.')
        pair = frame[['model_a', 'model_b', 'subset', 'metric']].copy()
        pair['subset'] = pair['subset'].map(normalize_subset)
        pair['metric'] = pair['metric'].map(normalize_metric)
        for column in ['value_a', 'value_b', 'difference', 'ci_low', 'ci_high', 'n', 'p_raw', 'p_adjusted']:
            pair[column] = numbers(choose_column(frame, [column]))
        for column, default in [('test_method', source.get('test_method', 'method not supplied')),
                                ('holm_family', source.get('holm_family', 'correction family not supplied')),
                                ('confidence', source.get('confidence', 0.95))]:
            pair[column] = choose_column(frame, [column], default)
        pair['dataset'], pair['source_file'] = source['dataset'], str(pair_path)
        pair['source_record'], pair['record_type'] = np.arange(1, len(pair)+1), 'comparison'
        pairs.append(pair)
    return result, pd.concat(pairs, ignore_index=True) if pairs else pd.DataFrame(), notices


# Native saved-statistics formats; these readers never evaluate predictions.
TASK_RESULT_FILES = {
    'kpuu': ('transfer_metrics.csv', 'paired_differences_with_pvalues.csv', 'statistics_manifest.json'),
    'herg': ('metrics_with_ci.csv', 'paired_comparisons.csv', 'analysis_manifest.json'),
}
KPUU_STATISTIC_METRICS = {
    'kendall_tau_b': 'kendall_tau', 'roc_auc_low_Kpuu': 'roc_auc',
    'average_precision_low_Kpuu': 'average_precision',
    # The fixed S4 PR AUC row uses the trapezoidal area; AP stays a separate metric.
    'pr_auc_trapezoidal_low_Kpuu': 'pr_auc_nonpenetrant',
    'accuracy_BBB_rule': 'accuracy', 'balanced_accuracy_BBB_rule': 'balanced_accuracy',
    'precision_low_Kpuu_BBB_rule': 'np_precision', 'recall_low_Kpuu_BBB_rule': 'np_recall',
    'f1_low_Kpuu_BBB_rule': 'np_f1', 'specificity_high_Kpuu_BBB_rule': 'specificity',
    'mcc_BBB_rule': 'mcc', 'cohens_kappa_BBB_rule': 'cohens_kappa',
}
HERG_STATISTIC_METRICS = {
    'nonblocker_specificity': 'nonblocker_recall', 'pr_auc_trapezoidal': 'pr_auc_trapezoid',
    'wrong_side_rate': 'misclassification_rate', 'wrong_side_distance': 'wrong_side_mean_distance',
}
HERG_STATISTIC_SUBSETS = {
    ('classification', 'all'): 'all_classification',
    ('classification', 'measured'): 'measured_classification',
    ('classification', 'screen_only'): 'classification_only',
    ('regression', 'measured'): 'measured_regression',
}


def load_task_statistics_source(source):
    """Read the exact outputs of 19_kpuu_statistics.py or 21_herg_statistics.py."""
    fmt, dataset = source['format'], source['dataset']
    metrics_name, pairs_name, manifest_name = TASK_RESULT_FILES[fmt]
    path = location(source, 'estimates_file', metrics_name)
    manifest_path = location(source, 'metadata_file', manifest_name)
    metadata = json.loads(manifest_path.read_text()) if manifest_path.is_file() else {}
    raw = pd.read_csv(path, low_memory=False)

    def require(frame, columns, origin):
        absent = set(columns) - set(frame)
        if absent:
            raise ValueError(f'{origin}: missing native {fmt} statistics columns {sorted(absent)}. Use the statistics script output directory, not a predictions/fold-summary file.')
        if frame.empty or frame[list(columns)].isna().any().any():
            raise ValueError(f'{origin}: empty statistics or missing model/metric/population identifiers.')

    def canonical(frame):
        frame = frame.copy()
        frame['source_metric'] = frame['metric']
        metric_map = KPUU_STATISTIC_METRICS if fmt == 'kpuu' else HERG_STATISTIC_METRICS
        frame['metric'] = frame.metric.map(lambda name: metric_name(source.get('metric_map', {}).get(name, metric_map.get(name, name))))
        if fmt == 'kpuu':
            frame['subset'] = 'all_classification'
        else:
            frame['source_subset'] = frame['subset']
            keys = list(zip(frame.family, frame.subset))
            unknown = set(keys) - set(HERG_STATISTIC_SUBSETS)
            if unknown:
                raise ValueError(f'{path}: unrecognized hERG family/subset combinations: {sorted(unknown)}.')
            frame['subset'] = [HERG_STATISTIC_SUBSETS[key] for key in keys]
        return frame

    identifiers = ['model', 'metric'] + (['family', 'subset'] if fmt == 'herg' else [])
    require(raw, identifiers, path)
    required_values = ['estimate', 'ci_low', 'ci_high'] + (['n_compounds'] if fmt == 'kpuu' else ['n_subset', 'n_metric'])
    absent = set(required_values) - set(raw)
    if absent:
        raise ValueError(f'{path}: missing saved statistics columns {sorted(absent)}.')
    result = canonical(raw)
    result['source_model'] = result['model']
    result['value'] = numbers(raw['estimate'])
    result['n'] = numbers(raw['n_compounds' if fmt == 'kpuu' else 'n_subset'])
    for column in ['ci_low', 'ci_high']:
        result[column] = numbers(result[column])
    if (result.ci_low > result.ci_high).any():
        raise ValueError(f'{path}: saved model-estimate CI bounds are reversed.')
    result['dataset'], result['format'] = dataset, fmt
    result['source_file'], result['source_record'] = str(path), np.arange(1, len(result) + 1)
    result['metadata_file'] = str(manifest_path) if manifest_path.is_file() else ''
    result['metric_model'], result['configured_reference'] = result['model'], source.get('reference', '')
    result['confidence'], result['record_type'] = .95, 'estimate'
    result['positive_class'] = 'low Kp,uu' if fmt == 'kpuu' else 'hERG blocker'
    result['distance_definition'] = ('Not applicable to this transfer-task table.' if fmt == 'kpuu' else
        "Mean absolute distance from the saved pIC50 threshold among each model's own misclassified compounds.")
    if fmt == 'herg':
        settings = metadata.get('settings', {})
        roles = source.get('model_roles', {'3': settings.get('reference', 'mse'), '4': settings.get('comparison', 'mtl')})
        if len(set(roles.values())) != len(roles):
            raise ValueError('hERG model_roles must assign different saved IDs to the publication roles.')
        result['publication_role'] = result.model.map({model: role for role, model in roles.items()})
    keys = ['subset', 'metric', 'model']
    if result.duplicated(keys).any():
        raise ValueError(f'{path}: duplicate native model/subset/metric estimates.')
    lookup = result.set_index(keys)

    p_column = source.get('p_column', 'p_holm')
    allowed = ['p_holm', 'p_holm_family'] if fmt == 'kpuu' else ['p_holm']
    if p_column not in allowed:
        raise ValueError(f'{dataset}: p_column must be one of {allowed}; raw p-values cannot be used as adjusted p-values.')
    paths = [Path(p) for p in source['pairwise_files']] if 'pairwise_files' in source else [Path(source['directory']) / pairs_name]
    pairs, notices = [], []
    for pair_path in paths:
        if not pair_path.is_file():
            notices.append(f'{dataset}: pairwise statistics missing ({pair_path}); estimates remain available.')
            continue
        saved = pd.read_csv(pair_path, low_memory=False)
        model_column = 'model' if fmt == 'kpuu' else 'comparison'
        require(saved, [model_column, 'reference', 'metric'] + (['family', 'subset'] if fmt == 'herg' else []), pair_path)
        fields = ['delta', p_column, 'p_value' if fmt == 'kpuu' else 'p_raw',
                  'n_compounds' if fmt == 'kpuu' else 'n_subset']
        fields += ['ci_low', 'ci_high', 'multiplicity_family'] if fmt == 'kpuu' else [
            'delta_ci_low', 'delta_ci_high', 'reference_estimate', 'comparison_estimate']
        absent = set(fields) - set(saved)
        if absent:
            raise ValueError(f'{pair_path}: missing native comparison columns {sorted(absent)}.')
        pair = canonical(saved)
        # Both scripts save delta = comparison/model minus reference, exactly A - B.
        pair['model_a'], pair['model_b'] = saved[model_column], saved['reference']
        pair['difference'] = numbers(saved['delta'])
        pair['ci_low'] = numbers(saved['ci_low' if fmt == 'kpuu' else 'delta_ci_low'])
        pair['ci_high'] = numbers(saved['ci_high' if fmt == 'kpuu' else 'delta_ci_high'])
        pair['n'] = numbers(saved['n_compounds' if fmt == 'kpuu' else 'n_subset'])
        pair['p_raw'] = numbers(saved['p_value' if fmt == 'kpuu' else 'p_raw'])
        pair['p_adjusted'], pair['p_value_column'] = numbers(saved[p_column]), p_column
        for side, native_value in [('a', 'comparison_estimate'), ('b', 'reference_estimate')]:
            index = pd.MultiIndex.from_arrays([pair.subset, pair.metric, pair['model_' + side]], names=keys)
            if not index.isin(lookup.index).all():
                raise ValueError(f'{pair_path}: comparison references estimates absent from {path}.')
            points = lookup.reindex(index)
            values = points.value.to_numpy(float)
            if fmt == 'herg' and not np.allclose(numbers(saved[native_value]), values, atol=1e-8, rtol=1e-6, equal_nan=True):
                raise ValueError(f'{pair_path}: saved {native_value} disagrees with the model estimates file.')
            pair['value_' + side] = values if fmt == 'kpuu' else numbers(saved[native_value])
            if not np.allclose(pair.n, points.n, atol=0, rtol=0, equal_nan=True):
                raise ValueError(f'{pair_path}: paired population sizes disagree with the estimates file.')
        if fmt == 'kpuu':
            if p_column == 'p_holm':
                pair['holm_family'] = 'all saved Kp,uu model-pair/metric comparisons in this analysis'
            else:
                pair['holm_family'] = 'all saved Kp,uu model-pair/metric comparisons within ' + saved.multiplicity_family.astype(str)
            pair['test_method'] = 'paired two-sided output-swap permutation test (score and binary call swapped together)'
        else:
            pair['holm_family'] = 'all saved hERG ' + saved.family.astype(str) + ' metric/subset comparisons in this analysis'
            pair['test_method'] = 'paired two-sided prediction-swap permutation test'
        pair['confidence'], pair['dataset'], pair['format'] = .95, dataset, fmt
        pair['source_file'], pair['source_record'] = str(pair_path), np.arange(1, len(pair) + 1)
        pair['record_type'] = 'comparison'
        pairs.append(pair)
    notices.append(f'{dataset}: read native {fmt} statistics; selected adjusted-p column {p_column}. Original model CIs, difference CIs and resampling diagnostics are retained in all_table_statistics.csv.')
    if fmt == 'kpuu':
        notices.append('Kp,uu: S4 PR AUC uses pr_auc_trapezoidal_low_Kpuu; average_precision_low_Kpuu is retained separately in the statistics export.')
    else:
        notices.append('hERG: S5 uses classification/all; S6 uses regression/measured. Measured classification and screen-only classification remain separate audit populations. Captions use n_subset, not class-specific n_metric.')
    return result, pd.concat(pairs, ignore_index=True) if pairs else pd.DataFrame(), notices


def load_result_source(source):
    directory, fmt, dataset = Path(source["directory"]), source["format"], source["dataset"]
    if fmt in TASK_RESULT_FILES:
        return load_task_statistics_source(source)
    if fmt == "long":
        return load_normalized_source(source)
    if fmt not in ["09", "11"]:
        raise ValueError("Source format must be 09, 11, long, kpuu, or herg.")
    default_metrics = "pooled_metrics_selected_models_long.csv" if fmt == "09" else "02_model_metrics.csv"
    estimates_path = location(source, "estimates_file", default_metrics)
    if not estimates_path.exists():
        if source.get("required", True):
            raise FileNotFoundError(f"{estimates_path} was not found. Set directory to your saved evaluation output folder.")
        return pd.DataFrame(), pd.DataFrame(), [f"{dataset}: optional input missing ({estimates_path})."]
    raw = pd.read_csv(estimates_path, low_memory=False)
    if fmt == "11":
        raw = raw.rename(columns=source.get("estimate_column_map", {}))
        if raw.columns.duplicated().any():
            raise ValueError(f"{estimates_path}: estimate_column_map creates duplicate metric columns.")
    metadata_path = location(source, "metadata_file", "analysis_metadata.json" if fmt == "09" else "05_evaluation_config.json")
    metadata = json.loads(metadata_path.read_text()) if metadata_path.exists() else {}
    confidence = source.get("confidence", metadata.get("confidence", 0.95))
    audit = read_optional(location(source, "audit_file", "selected_model_input_audit.csv")) if fmt == "09" else pd.DataFrame()
    if fmt == "09":
        required = {"model", "subset", "metric", "value"}
        if not required.issubset(raw):
            raise ValueError(f"{estimates_path} must be the LONG pooled metrics file; missing {required - set(raw)}.")
        estimates = raw[["model", "subset", "metric", "value"]].copy()
        estimates["n"] = numbers(choose_column(raw, ["n_samples", "n"]))
    else:
        available = [c for c in raw if metric_name(c) in ALL_METRICS]
        if "model" not in raw or not available:
            raise ValueError(f"{estimates_path}: no model or recognized metric columns.")
        if "subset" not in raw:
            raw["subset"] = "all"
        if "n" not in raw:
            raw["n"] = choose_column(raw, ["n_samples"])
        estimates = raw.melt(id_vars=["model", "subset", "n"], value_vars=available, var_name="metric", value_name="value")
        if source.get("estimate_column_map"):
            original_names = {new: old for old, new in source["estimate_column_map"].items()}
            estimates["source_metric"] = estimates.metric.map(lambda name: original_names.get(name, name))
    if source.get("source_note"):
        estimates["source_note"] = source["source_note"]
    estimates["subset"] = estimates.subset.map(subset_name)
    estimates["metric"] = estimates.metric.map(metric_name)
    estimates["value"] = numbers(estimates.value)
    estimates["dataset"], estimates["format"] = dataset, fmt
    estimates["source_file"], estimates["metric_model"] = str(estimates_path), estimates.model
    estimates["distance_definition"] = source.get("distance_definition", metadata.get("distance_definition", DISTANCE_09 if fmt == "09" else DISTANCE_11))
    estimates["configured_reference"] = source.get("reference", "")
    estimates["record_type"] = "estimate"
    default_pairs = "paired_two_sided_permutation_holm_by_pair_and_dataset.csv" if fmt == "09" else "04_pairwise_model_comparisons.csv"
    pair_paths = [Path(p) for p in source["pairwise_files"]] if "pairwise_files" in source else [directory / default_pairs]
    pairs, notices = [], []
    for path in pair_paths:
        if not path.exists():
            notices.append(f"{dataset}: pairwise file missing ({path}); unavailable annotations will be flagged.")
            continue
        try:
            frame = pd.read_csv(path, low_memory=False)
        except pd.errors.EmptyDataError:
            frame = pd.DataFrame()
        pairs.append(parse_pairs(frame, source, path, confidence))
    pairs = pd.concat(pairs, ignore_index=True) if pairs else pd.DataFrame()
    # Some external evaluators save regression point estimates only inside pairwise output.
    if fmt == "11" and not pairs.empty:
        candidates = []
        for _, pair in pairs[pairs.subset == "measured_regression"].iterrows():
            for model, value in [(pair.model_a, pair.value_a), (pair.model_b, pair.value_b)]:
                if pd.isna(value):
                    continue
                exists = (estimates.model == model) & (estimates.subset == pair.subset) & (estimates.metric == pair.metric)
                if exists.any():
                    continue
                candidates.append({"model": model, "subset": pair.subset, "metric": pair.metric, "value": value, "n": pair.n,
                                   "dataset": dataset, "format": fmt, "source_file": pair.source_file, "metric_model": model,
                                   "distance_definition": source.get("regression_distance_definition", DISTANCE_09),
                                   "configured_reference": source.get("reference", ""), "record_type": "estimate"})
        if candidates:
            supplemental = pd.DataFrame(candidates)
            for key, group in supplemental.groupby(["model", "subset", "metric"]):
                if not np.allclose(group.value, group.value.iloc[0], atol=1e-8, rtol=1e-6) or group.n.dropna().nunique() > 1:
                    raise ValueError(f"{dataset}/{key}: pair-specific estimates differ; provide a pooled estimate file for this population.")
            estimates = pd.concat([estimates, supplemental.drop_duplicates(["model", "subset", "metric"])], ignore_index=True)
    # Copy only point estimates explicitly declared identical in the evaluator's audit.
    if not audit.empty:
        copies = []
        for _, row in audit.iterrows():
            for column, metrics in [("regression_metrics_represented_by", REG_METRICS), ("score_metrics_represented_by", ["roc_auc", "pr_auc_nonpenetrant"])]:
                representative = row.get(column)
                if pd.isna(representative) or not str(representative).strip():
                    continue
                match = estimates[(estimates.model == representative) & estimates.metric.isin(metrics)].copy()
                if column == "regression_metrics_represented_by":
                    match = match[match.subset == "measured_regression"]
                existing = set(map(tuple, estimates.loc[estimates.model == row.model, ["subset", "metric"]].to_numpy()))
                match = match.loc[[tuple(x) not in existing for x in match[["subset", "metric"]].to_numpy()]]
                match["model"] = row.model
                copies.append(match)
        if copies:
            estimates = pd.concat([estimates, *copies], ignore_index=True)
        count_column = {"classification_only": "n_classification_only", "all_classification": "n_rows", "measured_regression": "n_measured", "measured_classification": "n_measured"}
        if audit.model.duplicated().any():
            raise ValueError(f"{dataset}: duplicate model rows in the audit file.")
        audit = audit.set_index("model")
        for i, row in estimates.iterrows():
            column = count_column.get(row.subset)
            if pd.isna(row.n) and row.model in audit.index and column in audit.columns:
                estimates.loc[i, "n"] = audit.loc[row.model, column]
    # Fill n from saved paired counts only when that model/subset has a single reported count.
    if not pairs.empty:
        for i, row in estimates[estimates.n.isna()].iterrows():
            counts = pairs.loc[(pairs.subset == row.subset) & ((pairs.model_a == row.model) | (pairs.model_b == row.model)), "n"].dropna().unique()
            if len(counts) == 1:
                estimates.loc[i, "n"] = counts[0]
    return estimates, pairs, notices

def difference_diagnostics(pairs):
    """Find inconsistencies beyond numerical precision, without changing any saved value."""
    if pairs.empty:
        return pd.DataFrame()
    values = pairs[["difference", "value_a", "value_b"]].to_numpy(dtype=float)
    finite = np.isfinite(values).all(axis=1)
    expected = pairs["value_a"] - pairs["value_b"]
    error = np.abs(pairs["difference"] - expected)
    # Subtraction can cancel two much larger, rounded component values. Scale the
    # tolerance to those components, not to their small difference (the old check).
    scale = np.maximum(np.abs(pairs["value_a"]), np.abs(pairs["value_b"]))
    tolerance = 1e-8 + 1e-6 * scale
    tolerance = tolerance.where(pairs["metric"] != "np_wrong_side_count", 1e-8)
    bad = finite & (error > tolerance)
    columns = [c for c in ["dataset", "subset", "model_a", "model_b", "metric", "source_file", "source_record", "value_a", "value_b", "difference", "ci_low", "ci_high", "p_adjusted"] if c in pairs]
    report = pairs.loc[bad, columns].copy()
    report["expected_a_minus_b"] = expected.loc[bad]
    report["absolute_mismatch"] = error.loc[bad]
    report["allowed_roundoff"] = tolerance.loc[bad]
    report["matches_b_minus_a"] = (np.abs(pairs.loc[bad, "difference"] + expected.loc[bad]) <= tolerance.loc[bad])
    return report

def check_results(estimates, pairs):
    if estimates.empty:
        raise ValueError("No saved estimates were loaded.")
    if estimates.duplicated(["dataset", "subset", "model", "metric"]).any():
        raise ValueError("Duplicate estimates: use one result source per dataset name.")
    if not pairs.empty:
        unordered = pairs.apply(lambda r: "\x1f".join(sorted([str(r.model_a), str(r.model_b)])), axis=1)
        if pairs.assign(pair=unordered).duplicated(["dataset", "subset", "pair", "metric"]).any():
            raise ValueError("Duplicate pair/metric results. Select one saved analysis per dataset; bootstrap/permutation outputs are not silently mixed.")
        finite = pairs.p_adjusted.dropna()
        if not finite.between(0, 1).all():
            raise ValueError("Adjusted p-values must be between 0 and 1.")
        if (pairs.ci_low > pairs.ci_high).any():
            raise ValueError("Saved CI lower bound exceeds its upper bound.")
        inconsistent = difference_diagnostics(pairs)
        if not inconsistent.empty:
            OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
            report_path = OUTPUT_DIR / "inconsistent_saved_differences.csv"
            inconsistent.to_csv(report_path, index=False)
            preview_columns = [c for c in ["dataset", "subset", "model_a", "model_b", "metric", "value_a", "value_b", "difference", "expected_a_minus_b", "absolute_mismatch"] if c in inconsistent]
            preview = inconsistent[preview_columns].head(10).to_string(index=False, float_format=lambda x: f"{x:.12g}")
            raise ValueError(f"{len(inconsistent)} saved difference(s) disagree with A − B beyond numerical roundoff.\n{preview}\nFull source records: {report_path.resolve()}\nSaved values, CIs, and p-values were not changed. Share this report to identify the source mismatch; a reversed CI must not be assumed from the difference alone.")

"""Direct auxiliary inputs; no model fitting, compound removal, or inferential tests."""
import ast
import glob
import re

auxiliary_audit = []
PARAMETER_LABELS = {
    'beta': 'Beta', 'margin_buffer': 'Margin buffer', 'learning_rate': 'Learning rate',
    'num_leaves': 'Number of leaves', 'max_depth': 'Maximum depth',
    'min_child_samples': 'Minimum child samples', 'subsample': 'Subsample',
    'colsample_bytree': 'Column samples by tree',
}
SEARCH_PARAMETER_ORDER = ['colsample_bytree', 'learning_rate', 'margin_buffer', 'beta',
                          'max_depth', 'num_leaves', 'min_child_samples', 'subsample']
PARAMETER_DESCRIPTIONS = {
    'colsample_bytree': 'Fraction of features sampled for each tree',
    'learning_rate': 'Learning rate', 'margin_buffer': 'Custom-loss margin buffer',
    'beta': 'Classification-loss weight', 'max_depth': 'Maximum tree depth',
    'num_leaves': 'Maximum number of leaves per tree',
    'min_child_samples': 'Minimum samples in a leaf',
    'subsample': 'Fraction of training samples used for each tree',
}

class AuxiliaryUnavailable(Exception):
    """An optional source has not been supplied; retain the publication number."""

def auxiliary_path(value, base=None):
    if value is None:
        raise AuxiliaryUnavailable('Input path is not configured.')
    path = Path(value).expanduser()
    return Path(base) / path if base is not None and not path.is_absolute() else path

def auxiliary_record(key, field, value, source='', fold='', detail=''):
    auxiliary_audit.append(dict(table=key, field=field, fold=fold, value=value,
                                source_file=str(source), detail=detail))

def auxiliary_issue(key, field, issue):
    report('', '', key, '', field, issue)

def auxiliary_token(value):
    if pd.isna(value):
        return None
    text = str(value).strip()
    try:
        numeric = float(text)
        if np.isfinite(numeric):
            return str(int(numeric)) if numeric.is_integer() else str(numeric)
    except ValueError:
        pass
    return text.casefold()

def auxiliary_json_value(data, key):
    for part in key.split('.') if key else []:
        data = data[int(part)] if isinstance(data, list) else data[part]
    return data

def read_auxiliary_frame(source, base=None):
    """Read explicit CSV/TSV/JSON/parquet/SDF paths. JSON code is never executed."""
    source = {'path': source} if isinstance(source, (str, Path)) else source
    requested = source.get('path')
    if requested is None:
        raise AuxiliaryUnavailable('Set the source path in the first code cell.')
    requested = requested if isinstance(requested, (list, tuple)) else [requested]
    paths = []
    for value in requested:
        pattern = str(auxiliary_path(value, base))
        matches = sorted(glob.glob(pattern, recursive=True)) if glob.has_magic(pattern) else [pattern]
        if not matches:
            raise AuxiliaryUnavailable(f'No input matches {pattern}.')
        paths.extend(Path(p) for p in matches)
    if len(set(p.resolve() for p in paths)) != len(paths):
        raise ValueError('The same auxiliary input file was selected more than once.')
    frames = []
    for path in paths:
        if not path.exists():
            raise AuxiliaryUnavailable(f'Input file does not exist: {path}')
        if not path.is_file():
            raise ValueError(f'{path}: specify the dataset file, not its containing directory.')
        suffix = path.suffix.lower()
        if suffix == '.json':
            data = auxiliary_json_value(json.loads(path.read_text()), source.get('records_key'))
            if source.get('fold_keyed'):
                if not isinstance(data, dict):
                    raise ValueError(f'{path}: fold_keyed=True requires a JSON object keyed by fold.')
                data = [dict({'fold': fold}, **(value if isinstance(value, dict) else {'value': value}))
                        for fold, value in data.items()]
            frame = pd.json_normalize(data if isinstance(data, list) else [data])
        elif suffix in ['.sdf', '.sd']:
            from rdkit import Chem
            rows = []
            for index, molecule in enumerate(Chem.SDMolSupplier(str(path), sanitize=False, removeHs=False)):
                if molecule is None:
                    raise ValueError(f'{path}: unreadable SDF record {index+1}; no records were discarded.')
                rows.append(molecule.GetPropsAsDict(autoConvertStrings=False))
            frame = pd.DataFrame(rows)
        elif suffix == '.parquet':
            frame = pd.read_parquet(path)
        else:
            options = dict(source.get('read_csv_kwargs', {}))
            options.setdefault('sep', '\t' if suffix == '.tsv' else ',')
            options.setdefault('dtype', object)  # Preserve identifiers such as 0012.
            frame = pd.read_csv(path, **options)
        frame = frame.rename(columns=source.get('column_map', {}))
        if not frame.columns.is_unique or '_source_file' in frame:
            raise ValueError(f'{path}: duplicate columns or reserved _source_file column.')
        if frame.empty:
            raise ValueError(f'{path}: input contains no data rows.')
        frame['_source_file'] = str(path)
        if source.get('fold') is not None and source.get('fold_column', 'fold') not in frame:
            frame[source.get('fold_column', 'fold')] = source['fold']
        frames.append(frame)
    if not frames:
        raise AuxiliaryUnavailable('No auxiliary input files were selected.')
    return pd.concat(frames, ignore_index=True)

def inventory_inputs(directory=DATASET_DIR, recursive=False):
    """List candidate dataset filenames and CSV headers; never combine their rows."""
    root = Path(directory).expanduser()
    paths = sorted(root.rglob('*') if recursive else root.glob('*'))
    records = []
    for path in paths:
        if path.is_file() and path.suffix.lower() in ['.csv', '.tsv', '.sdf', '.sd', '.parquet', '.json']:
            columns = ''
            if path.suffix.lower() in ['.csv', '.tsv']:
                try:
                    columns = ', '.join(pd.read_csv(path, sep='\t' if path.suffix.lower()=='.tsv' else ',', nrows=0).columns)
                except (ValueError, OSError, pd.errors.ParserError) as error:
                    columns = str(error)
            records.append({'file': str(path), 'columns': columns})
    return pd.DataFrame(records, columns=['file', 'columns'])

def find_threshold_sources(directory=Path('.')):
    """Find candidate CSV headers/JSON keys. Candidates are never used automatically."""
    root, records = Path(directory).expanduser(), []
    for path in sorted(root.rglob('*')):
        if not path.is_file() or any(part in {'.git', '.ipynb_checkpoints'} for part in path.parts):
            continue
        try:
            if path.suffix.lower() in ['.csv', '.tsv']:
                names = list(pd.read_csv(path, sep='\t' if path.suffix.lower()=='.tsv' else ',', nrows=0).columns)
            elif path.suffix.lower() == '.json':
                def keys(value, prefix=''):
                    if isinstance(value, dict):
                        for key, child in value.items():
                            full = f'{prefix}.{key}' if prefix else str(key)
                            yield full
                            yield from keys(child, full)
                    elif isinstance(value, list):
                        for child in value:
                            yield from keys(child, prefix)
                names = list(keys(json.loads(path.read_text())))
            else:
                continue
            hits = sorted({name for name in names if 'threshold' in str(name).lower()})
            if hits:
                records.append({'file': str(path), 'fields': ', '.join(hits)})
        except (ValueError, OSError, UnicodeError, pd.errors.ParserError):
            continue
    return pd.DataFrame(records, columns=['file', 'fields'])

def build_dataset_counts(key='main_07'):
    if not any(source.get('path') is not None for source in DATASET_COUNT_SOURCES.values()):
        raise AuxiliaryUnavailable('Set the original/training/external dataset paths.')
    records = []
    for label in DATASET_PAPER_ROWS:
        source = dict(DATASET_COUNT_SOURCES.get(label,{}))
        blank = {'Dataset':label,'BBB penetrant (BBB+)':'NA','BBB nonpenetrant (BBB-)':'NA','Total':'NA'}
        try:
            path = auxiliary_path(source.get('path'),DATASET_DIR)
            if path.suffix.lower() in ['.csv','.tsv'] and path.is_file():
                options = dict(source.get('read_csv_kwargs',{}))
                sep = options.get('sep','\t' if path.suffix.lower()=='.tsv' else ',')
                headers = list(pd.read_csv(path,sep=sep,nrows=0).columns)
                if source.get('class_column') is None:
                    candidates = [c for c in ['bbb_class','Class','BBB+/BBB-'] if c in headers]
                    if len(candidates)==1:
                        source['class_column'] = candidates[0]
                if not source.get('no_measured_logbb') and not source.get('measured_flag_column'):
                    flags = [c for c in headers if c.lower() in ['has_logbb','has_measured_logbb','is_measured']]
                    if len(flags)==1:
                        source['measured_flag_column'] = flags[0]
                if not source.get('no_measured_logbb') and not source.get('measured_flag_column') and not source.get('measured_logbb_column'):
                    observed = [c for c in headers if c.lower() in ['logbb','true_logbb']]
                    if len(observed)==1:
                        source['measured_logbb_column'] = observed[0]
                needed = [source.get(name) for name in ['class_column','id_column','measured_flag_column','measured_logbb_column']]
                if not source.get('class_column'):
                    needed += [c for c in ['bbb_class','Class','BBB+/BBB-'] if c in headers]
                if not options.get('usecols'):
                    options['usecols'] = list(dict.fromkeys(c for c in needed if c))
                source['read_csv_kwargs'] = options
            frame = read_auxiliary_frame(source,DATASET_DIR)
        except AuxiliaryUnavailable as error:
            auxiliary_issue(key,label,str(error))
            records.append(blank)
            continue
        identifier = source.get('id_column')
        if identifier and (frame[identifier].isna().any() or frame[identifier].duplicated().any()):
            raise ValueError(f'{label}: missing/duplicate IDs; no rows were discarded.')
        column = source.get('class_column')
        if not column or column not in frame:
            raise ValueError(f'{label}: could not identify the saved BBB-class column.')
        mapping = source.get('class_map',{0:'penetrant',1:'nonpenetrant','BBB+':'penetrant','BBB-':'nonpenetrant'})
        mapping = {auxiliary_token(value):name for value,name in mapping.items()}
        classes = frame[column].map(auxiliary_token).map(mapping)
        if classes.isna().any() or not set(mapping.values()).issubset({'penetrant','nonpenetrant'}):
            raise ValueError(f'{label}: unmapped/missing class labels; configure class_map.')
        flag,logbb = source.get('measured_flag_column'),source.get('measured_logbb_column')
        if source.get('no_measured_logbb'):
            measured = pd.Series(False,index=frame.index)
        elif flag:
            positive = {auxiliary_token(v) for v in source.get('measured_values',[1,True])}
            negative = {auxiliary_token(v) for v in source.get('unmeasured_values',[0,False])}
            tokens = frame[flag].map(auxiliary_token)
            if positive & negative or not tokens.isin(positive|negative).all():
                raise ValueError(f'{label}: measurement availability flags are missing or unknown.')
            measured = tokens.isin(positive)
        elif logbb:
            observed = pd.to_numeric(frame[logbb],errors='raise')
            if (observed.notna() & ~np.isfinite(observed)).any():
                raise ValueError(f'{label}: nonfinite measured logBB values.')
            measured = observed.notna()
        else:
            measured = None
            auxiliary_issue(key,label,'Experimental logBB availability is not defined; set the original observation/availability column.')
        cells = {}
        for category in ['penetrant','nonpenetrant','total']:
            mask = classes.eq(category) if category!='total' else pd.Series(True,index=frame.index)
            count = int(mask.sum())
            observed = int((mask & measured).sum()) if measured is not None else None
            cells[category] = f'{count} ({observed if observed is not None else "NA"})'
            auxiliary_record(key,f'{label}/{category}/count',count,path,detail=f'class_column={column}')
            auxiliary_record(key,f'{label}/{category}/measured_count',observed,path,detail=f'availability_flag={flag}; observation={logbb}; explicit_none={source.get("no_measured_logbb",False)}')
        records.append({'Dataset':label,'BBB penetrant (BBB+)':cells['penetrant'],
                        'BBB nonpenetrant (BBB-)':cells['nonpenetrant'],'Total':cells['total']})
    if all(row['Total']=='NA' for row in records):
        raise AuxiliaryUnavailable('None of the dataset files are available; check PROJECT_DIR/DATASET_DIR.')
    return pd.DataFrame(records),'Counts calculated from the supplied datasets. Parentheses contain counts with experimental logBB, using saved availability flags when present. Missing original datasets remain NA; no rows were filtered or duplicated to fill them.'

def descriptor_features(entry, family):
    source = dict(entry) if isinstance(entry, dict) else {'path': entry}
    source.setdefault('exclude_columns', family.get('exclude_columns', []))
    source.setdefault('id_column', family.get('id_column'))
    path = auxiliary_path(source.get('path'), DATASET_DIR)
    if not path.is_file():
        raise AuxiliaryUnavailable(f'Descriptor input is missing: {path}')
    if source.get('format') == 'feature_list':
        if path.suffix.lower() == '.json':
            names = auxiliary_json_value(json.loads(path.read_text()), source.get('records_key'))
        elif path.suffix.lower() == '.txt':
            names = [line.strip() for line in path.read_text().splitlines() if line.strip()]
        else:
            frame = read_auxiliary_frame(source, DATASET_DIR).drop(columns='_source_file')
            column = source.get('feature_column', 'feature' if 'feature' in frame else frame.columns[0] if len(frame.columns)==1 else None)
            if column is None:
                raise ValueError(f'{path}: set feature_column for the feature-list CSV.')
            names = frame[column].tolist()
        if not isinstance(names, list) or not names or any(pd.isna(v) or not str(v).strip() for v in names):
            raise ValueError(f'{path}: expected a nonempty list of feature names.')
        names = [str(v) for v in names]
        if len(set(names)) != len(names):
            raise ValueError(f'{path}: repeated feature names.')
        family_name=source.get('family')
        if family_name in ['ECFP','MACCS']:
            names=[name for name in names if name.startswith(family_name+'_')]
        elif family_name=='Mordred':
            names=[name for name in names if not name.startswith(('ECFP_','MACCS_'))]
        if not names:
            raise ValueError(f'{path}: no feature names for family {family_name}.')
        return names, None, path
    frame = read_auxiliary_frame(source, DATASET_DIR).drop(columns='_source_file')
    excluded = set(source['exclude_columns']) | ({source['id_column']} if source['id_column'] else set())
    names = list(source.get('columns') or [c for c in frame if c not in excluded])
    if source.get('prefix') is not None:
        names = [c for c in names if c.startswith(source['prefix'])]
    family_name=source.get('family')
    if family_name in ['ECFP','MACCS']:
        names=[name for name in names if name.startswith(family_name+'_')]
    elif family_name=='Mordred':
        names=[name for name in names if not name.startswith(('ECFP_','MACCS_'))]
    if not names or len(set(names)) != len(names) or not set(names).issubset(frame):
        raise ValueError(f'{path}: select nonempty, unique descriptor columns using columns/prefix/exclude_columns.')
    return names, frame[names], path

def build_descriptor_counts(key='supplementary_07'):
    if not any(s.get('before') is not None or s.get('after') is not None for s in DESCRIPTOR_COUNT_SOURCES.values()):
        raise AuxiliaryUnavailable('Set before/after matrices or feature lists in DESCRIPTOR_COUNT_SOURCES.')
    before_counts, after_counts = {}, {}
    for family in DESCRIPTOR_PAPER_FAMILIES:
        source = DESCRIPTOR_COUNT_SOURCES.get(family,{})
        before, matrix = None, None
        for stage, counts in [('before', before_counts), ('after', after_counts)]:
            try:
                entry = source.get(stage)
                if stage == 'after' and entry is None and source.get('derive_after'):
                    if before is None:
                        raise AuxiliaryUnavailable(f'{family}: original descriptor matrix is unavailable; after-selection count cannot be derived.')
                    if source['derive_after'] != 'finite_all_rows' or matrix is None:
                        raise ValueError(f'{family}: derive_after="finite_all_rows" needs the original descriptor matrix.')
                    numeric = matrix.apply(pd.to_numeric, errors='coerce')
                    available = np.isfinite(numeric).all(axis=0)
                    if source.get('drop_constant', False):
                        available &= numeric.nunique(dropna=False) > 1
                    names = list(numeric.columns[available])
                    path = before_path
                    method = f'finite for all saved rows; drop_constant={source.get("drop_constant",False)}'
                else:
                    names, selected, path = descriptor_features(entry, source)
                    method = 'counted saved descriptor columns/feature names'
                    if stage == 'before':
                        before, matrix, before_path = names, selected, path
                if stage == 'after' and before is not None and not set(names).issubset(before):
                    raise ValueError(f'{family}: retained feature names are absent from the original feature set.')
                counts[family] = len(names)
                auxiliary_record(key, f'{family}/{stage}', len(names), path, detail=method)
            except AuxiliaryUnavailable as error:
                counts[family] = np.nan
                auxiliary_issue(key, f'{family}/{stage}', str(error))
    def row(label, counts):
        result = {'Descriptors': label, **{name: number(value, integer=True) for name, value in counts.items()}}
        result['Total'] = number(sum(counts.values()), integer=True) if all(pd.notna(v) for v in counts.values()) else 'NA'
        return result
    return pd.DataFrame([row('Available descriptors', before_counts),
                         row('Descriptors that were available for all molecules in the dataset', after_counts)]), 'Counts use the configured original and retained feature matrices/lists. Derived availability counts, when explicitly enabled, use finite values for every saved row; no compound rows are removed.'

def training_source_trees(source):
    path = auxiliary_path(source.get('path'))
    if not path.is_file():
        raise AuxiliaryUnavailable(f'Training source is missing: {path}')
    if path.suffix.lower() == '.ipynb':
        texts = [''.join(cell['source']) for cell in json.loads(path.read_text())['cells'] if cell['cell_type']=='code']
    else:
        texts = [path.read_text()]
    trees = []
    for text in texts:
        if text.lstrip().startswith('%%'):
            continue
        clean = '\n'.join('' if line.lstrip().startswith(('%', '!')) else line for line in text.splitlines())
        try:
            trees.append(ast.parse(clean))
        except SyntaxError as error:
            raise ValueError(f'{path}: could not parse the training source for search spaces: {error}') from error
    return path, trees

def build_search_spaces(key='supplementary_08', source=None):
    source = SEARCH_SPACE_SOURCE if source is None else source
    spaces = dict(source.get('spaces',{}))
    path = 'explicit configuration'
    if source.get('path') is not None:
        path, trees = training_source_trees(source)
        constants = {}
        for tree in trees:
            for node in tree.body:
                if isinstance(node, ast.Assign):
                    try:
                        value = ast.literal_eval(node.value)
                        for target in node.targets:
                            if isinstance(target, ast.Name):
                                constants.setdefault(target.id, []).append(value)
                    except (ValueError, TypeError):
                        pass
        def literal(node):
            if isinstance(node, ast.Name) and node.id in constants:
                values = constants[node.id]
                if all(value == values[0] for value in values):
                    return values[0]
            return ast.literal_eval(node)
        roots = trees
        if source.get('function'):
            roots = [node for tree in trees for node in ast.walk(tree)
                     if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)) and node.name == source['function']]
            if len(roots) != 1:
                raise ValueError(f'{path}: function must identify one objective; found {len(roots)}.')
        found = {}
        for root in roots:
            for node in ast.walk(root):
                if not isinstance(node, ast.Call) or not isinstance(node.func, ast.Attribute) or not node.func.attr.startswith('suggest_'):
                    continue
                arguments = {kw.arg: kw.value for kw in node.keywords}
                name = None
                try:
                    name = literal(node.args[0] if node.args else arguments['name'])
                    if not isinstance(name, str):
                        continue
                    kind = node.func.attr
                    if kind == 'suggest_categorical':
                        choices = literal(node.args[1] if len(node.args)>1 else arguments['choices'])
                        text = ', '.join(map(str, choices))
                    elif kind in ['suggest_float', 'suggest_int', 'suggest_uniform', 'suggest_loguniform', 'suggest_discrete_uniform']:
                        low = literal(node.args[1] if len(node.args)>1 else arguments['low'])
                        high = literal(node.args[2] if len(node.args)>2 else arguments['high'])
                        text = f'{low:g}-{high:g}'
                        if kind == 'suggest_loguniform' or ('log' in arguments and literal(arguments['log'])):
                            text += ' (log scale)'
                        step = arguments.get('step', arguments.get('q'))
                        if step is None and kind == 'suggest_discrete_uniform' and len(node.args)>3:
                            step = node.args[3]
                        if step is not None:
                            text += f'; step={literal(step):g}'
                    else:
                        continue
                except (ValueError, TypeError, KeyError, IndexError):
                    if name in SEARCH_PARAMETER_ORDER and name not in spaces:
                        auxiliary_issue(key, name, 'Dynamic search-space expression cannot be resolved without executing training. Supply SEARCH_SPACE_SOURCE["spaces"] for this parameter.')
                    continue
                if name in found and found[name] != text and name not in spaces:
                    raise ValueError(f'{path}: multiple search spaces for {name}; select function or provide an explicit spaces entry.')
                found[name] = text
        spaces = {**found, **spaces}
    if not spaces and source.get('path') is None:
        raise AuxiliaryUnavailable('Set SEARCH_SPACE_SOURCE.path to training .py/.ipynb or provide explicit spaces.')
    rows = []
    for name in source.get('parameters', SEARCH_PARAMETER_ORDER):
        value = spaces.get(name, 'NA')
        if value == 'NA':
            auxiliary_issue(key, name, 'Search space absent from configured training source/explicit definitions.')
        auxiliary_record(key, name, value, path)
        label = 'Custom loss function hyperparameter: '+name if name in ['beta', 'margin_buffer'] else name
        rows.append({'Hyperparameter Name': label, 'Description': source.get('descriptions', {}).get(name, PARAMETER_DESCRIPTIONS.get(name, name)), 'Search Space': value})
    return pd.DataFrame(rows), 'Search spaces extracted from training code without executing it, with any explicitly configured definitions retained. Dynamic or absent ranges remain NA.'

def fold_source_frame(source, expected_subset=None):
    if source.get('fold_base','auto') not in [0,1,'auto']:
        raise ValueError('fold_base must be 0, 1 or auto.')
    frame = canonical_fold_columns(read_auxiliary_frame(source),source)
    model_column = source.get('model_column', 'model')
    if model_column in frame:
        models = frame[model_column].dropna().unique()
        model = source.get('model')
        if model is None:
            model = paper_auxiliary_model(models,source)
        if model is not None:
            frame = frame[frame[model_column] == model].copy()
    selected = source.get('selected_column')
    if selected:
        token = auxiliary_token(source.get('selected_value', True))
        frame = frame[frame[selected].map(auxiliary_token) == token].copy()
    if expected_subset:
        if source.get('positive_class') not in [None, 'nonpenetrant']:
            raise ValueError('BBB classification tables require nonpenetrants as the positive class.')
        declared = source.get('subset')
        if declared is not None and subset_name(declared) != expected_subset:
            raise ValueError(f'Expected {expected_subset}, but configured subset={declared}.')
        column = source.get('subset_column', 'subset')
        if column in frame:
            actual = frame[column].map(lambda value: subset_name(source.get('subset_map', {}).get(value, value)))
            frame = frame[actual == expected_subset].copy()
        elif declared is None:
            raise ValueError(f'No subset column: explicitly declare subset="{expected_subset}" in this source.')
    if frame.empty:
        raise ValueError('No saved records match the configured model/selection/subset.')
    column = source.get('fold_column', 'fold')
    if column in frame:
        folds = frame[column]
    else:
        folds = frame._source_file.map(lambda p: (re.search(r'(?:outer[_-]?)?fold[_-]?(\d+)', str(p), re.I) or [None, None])[1])
    def parse_fold(value):
        if pd.isna(value):
            raise ValueError('Missing fold ID. Set fold_column/fold or use filenames containing fold_1, etc.')
        match = re.fullmatch(r'(?:outer[ _-]*)?fold[ _-]*(\d+)', str(value), re.I)
        numeric = float(match[1] if match else value)
        if not np.isfinite(numeric) or not numeric.is_integer():
            raise ValueError(f'Invalid fold ID: {value}')
        return int(numeric)
    frame['_fold'] = folds.map(parse_fold)
    base=source.get('fold_base','auto')
    ids=set(frame._fold)
    if base=='auto':
        base=0 if 0 in ids and ids.issubset(set(range(5))) else 1
    if base==0:
        frame['_fold'] += 1
    if not frame._fold.isin(EXPECTED_FOLDS).all():
        raise ValueError(f'Unexpected folds {sorted(frame._fold.unique())}; expected {EXPECTED_FOLDS}. Set fold_base=0 for zero-based IDs.')
    return frame

def saved_fold_number(frame, column, fold, key, field):
    rows = frame[frame._fold == fold]
    if column is None or column not in rows or rows.empty:
        auxiliary_issue(key, f'{field}/fold_{fold}', 'Saved per-fold value unavailable.')
        return np.nan
    values = pd.to_numeric(rows[column], errors='raise').dropna().unique()
    if len(values) > 1:
        raise ValueError(f'{key}/{field}/fold {fold}: conflicting values. Select one saved model/run/trial.')
    if not len(values) or not np.isfinite(values[0]):
        auxiliary_issue(key, f'{field}/fold_{fold}', 'Saved per-fold value undefined or unavailable.')
        return np.nan
    value = float(values[0])
    auxiliary_record(key, field, value, '; '.join(rows._source_file.unique()), fold)
    return value

def build_hyperparameters(key='supplementary_09', source=None):
    source = HYPERPARAMETER_SOURCE if source is None else source
    frame = expanded_parameter_frame(fold_source_frame(source),source)
    rows = []
    for parameter, label in PARAMETER_LABELS.items():
        candidates = [parameter, 'params_'+parameter, 'params.'+parameter, 'best_params.'+parameter]
        configured = source.get('parameter_columns', {}).get(parameter)
        available = [c for c in candidates if c in frame]
        if configured is None and len(available) > 1:
            raise ValueError(f'{parameter}: multiple parameter columns; set parameter_columns explicitly.')
        column = configured or (available[0] if available else None)
        selected_frame = frame
        if {'parameter','value'}.issubset(frame):
            selected_frame = frame[frame.parameter.map(lambda value: str(value).removeprefix('params_').removeprefix('best_params.'))==parameter]
            column = 'value'
        integer = parameter in ['num_leaves', 'max_depth', 'min_child_samples']
        row = {'Hyperparameter': label}
        for fold in EXPECTED_FOLDS:
            row[f'Fold {fold}'] = number(saved_fold_number(selected_frame, column, fold, key, parameter), integer=integer, trim=parameter in ['beta','margin_buffer'])
        rows.append(row)
    return pd.DataFrame(rows), 'Selected saved hyperparameters by outer fold. No tuning was rerun. Map selected_column when reading trial histories.'

def auxiliary_metric_name(value, source):
    value = source.get('metric_map', {}).get(value, value)
    aliases = {'balanced accuracy':'balanced_accuracy', 'precision':'np_precision', 'recall':'np_recall',
               'f1':'np_f1', 'roc auc':'roc_auc', 'pr auc':'pr_auc_nonpenetrant',
               'pcc':'pearson_r', 'r²':'r2', 'cohen’s kappa':'cohens_kappa', "cohen's kappa":'cohens_kappa'}
    lower = str(value).lower()
    return metric_name(aliases.get(lower, lower))

def build_fold_metrics(key, source=None):
    source = PER_FOLD_SOURCES[key] if source is None else source
    regression = key == 'supplementary_10'
    population = 'measured_regression' if regression else 'all_classification'
    frame = fold_source_frame(source,population)
    metrics = (['mse','mde','mae','pearson_r','r2'] if regression else
               ['balanced_accuracy','np_precision','np_recall','roc_auc','pr_auc_nonpenetrant','cohens_kappa','mcc','np_f1'])
    labels = dict(mse='MSE',mde='MDE',mae='MAE',pearson_r='PCC',r2='R2',balanced_accuracy='Balanced Accuracy',
                  np_precision='Precision',np_recall='Recall',roc_auc='ROC AUC',pr_auc_nonpenetrant='PR AUC',
                  cohens_kappa='kappa',mcc='MCC',np_f1='F1')
    # Combined exports contain metric/value columns for some models and wide
    # columns for others. A blank metric/value column is not a long record.
    parts = []
    long_rows = pd.Series(False,index=frame.index)
    if {'metric','value'}.issubset(frame):
        long_rows = frame.metric.notna() & frame.metric.astype(str).str.strip().ne('')
        saved = frame.loc[long_rows,['_fold','_source_file','metric','value']].copy()
        saved['metric'] = saved.metric.map(lambda value: auxiliary_metric_name(value,source))
        parts.append(saved[saved.metric.isin(metrics)])
    wide = frame.loc[~long_rows]
    columns = [c for c in wide if auxiliary_metric_name(c,source) in metrics]
    if columns and not wide.empty:
        saved = wide[['_fold','_source_file']+columns].melt(id_vars=['_fold','_source_file'],value_vars=columns,var_name='metric',value_name='value')
        saved['metric'] = saved.metric.map(lambda value: auxiliary_metric_name(value,source))
        parts.append(saved)
    long = pd.concat(parts,ignore_index=True) if parts else pd.DataFrame(columns=['_fold','_source_file','metric','value'])
    values = {metric:[saved_fold_number(long[long.metric==metric],'value',fold,key,metric) for fold in EXPECTED_FOLDS] for metric in metrics}
    rows = [{'Fold':f'Fold {fold}',**{labels[m]:number(values[m][index]) for m in metrics}} for index,fold in enumerate(EXPECTED_FOLDS)]
    summary = {'Fold':'Average ± SD'}
    for metric in metrics:
        array = np.asarray(values[metric])
        if np.isfinite(array).all():
            mean,sd = float(array.mean()),float(array.std(ddof=FOLD_SD_DDOF))
            summary[labels[metric]] = f'{number(mean,trim=True)} ± {number(sd,trim=True)}'
            auxiliary_record(key,metric+'/fold_mean',mean,detail='Unweighted mean of saved unrounded outer-fold metrics.')
            auxiliary_record(key,metric+'/fold_sd',sd,detail=f'ddof={FOLD_SD_DDOF}')
        else:
            summary[labels[metric]] = 'NA'
    return pd.DataFrame(rows+[summary]),f'Model 4 saved per-fold metrics for {population}. Fold IDs 0-4 are displayed as folds 1-5. The final row is the unweighted mean ± sample SD (ddof={FOLD_SD_DDOF}); incomplete metrics have no summary.'

def build_thresholds(key='supplementary_12', source=None):
    source = {**(THRESHOLD_SOURCE if source is None else source),'paper_role':'5'}
    frame = fold_source_frame(source)
    column = source.get('value_column')
    if column is None:
        preferred = ['tuned_threshold','selected_threshold','classification_threshold','best_threshold','inner_tuned_threshold']
        candidates = [c for c in preferred if c in frame and pd.to_numeric(frame[c],errors='coerce').notna().any()]
        if not candidates:
            candidates = [c for c in frame if 'threshold' in str(c).lower()
                          and not any(word in str(c).lower() for word in ['fixed','native','selection'])
                          and pd.to_numeric(frame[c],errors='coerce').notna().any()]
        if not candidates:
            raise ValueError('No saved selected numeric threshold field was found.')
        column = candidates[0]
        for candidate in candidates[1:]:
            a,b = pd.to_numeric(frame[column],errors='coerce'),pd.to_numeric(frame[candidate],errors='coerce')
            if not np.allclose(a,b,equal_nan=True,atol=1e-12,rtol=1e-10):
                raise ValueError(f'Saved selected-threshold fields disagree: {candidates}; set value_column.')
    rows = [{'Fold':f'Fold {fold}','Chosen threshold':number(saved_fold_number(frame,column,fold,key,'chosen_threshold'))} for fold in EXPECTED_FOLDS]
    return pd.DataFrame(rows),'Original inner-OOF selected thresholds read from saved model metadata. No cutoff selection or model fitting was rerun.'



"""Fixed paper layouts and compatibility routing for raw auxiliary CSVs."""
DATASET_PAPER_ROWS = ['B3DB', 'Data set used', 'External validation set']
DESCRIPTOR_PAPER_FAMILIES = ['Mordred', 'ECFP', 'MACCS']

def publication_layout(item):
    key = item['key']
    folds = [f'Fold {fold}' for fold in range(1, 6)]
    auxiliary = {
        'main_07': (['Dataset', 'BBB penetrant (BBB+)', 'BBB nonpenetrant (BBB-)', 'Total'], DATASET_PAPER_ROWS),
        'supplementary_07': (['Descriptors', 'Mordred', 'ECFP', 'MACCS', 'Total'],
                             ['Available descriptors', 'Descriptors that were available for all molecules in the dataset']),
        'supplementary_08': (['Hyperparameter Name', 'Description', 'Search Space'],
                             [('Custom loss function hyperparameter: '+p) if p in ['beta','margin_buffer'] else p for p in SEARCH_PARAMETER_ORDER]),
        'supplementary_09': (['Hyperparameter']+folds, list(PARAMETER_LABELS.values())),
        'supplementary_10': (['Fold','MSE','MDE','MAE','PCC','R2'], folds+['Average ± SD']),
        'supplementary_11': (['Fold','Balanced Accuracy','Precision','Recall','ROC AUC','PR AUC','kappa','MCC','F1'], folds+['Average ± SD']),
        'supplementary_12': (['Fold','Chosen threshold'], folds),
    }
    if key in auxiliary:
        return auxiliary[key]
    if item['style'] != 'auxiliary':
        first = 'Metric' if item['style'] in ['herg','kpuu'] else 'Model'
        columns = [first]+[column_label(role,item) for role in item['roles']]
        if item['style']=='herg':
            columns.append('Adjusted p-value')
        return columns, [metric_label(metric,item) for metric in item['metrics']]
    return None

def publication_token(value):
    return re.sub(r'\s+', ' ', str(value).replace('²','2').replace('−','-')).strip().casefold()

def align_publication_table(table, item, source=''):
    layout = publication_layout(item)
    if layout is None:
        return table
    columns, rows = layout
    # Recognize equivalent header typography; never use unrelated columns as table data.
    rename = {}
    for expected in columns:
        matches = [actual for actual in table if publication_token(actual)==publication_token(expected)]
        if len(matches)>1:
            raise ValueError(f'{item["key"]}: ambiguous presentation column {expected}.')
        if matches:
            rename[matches[0]] = expected
    table = table.rename(columns=rename).copy()
    first = columns[0]
    if first not in table:
        raise ValueError(f'{item["key"]}: finished table must contain {first!r}; raw analysis files must use the source reader.')
    if table[first].map(publication_token).duplicated().any():
        raise ValueError(f'{item["key"]}: repeated presentation row labels; select the intended saved model/subset through the raw reader.')
    actual = {publication_token(value): index for index,value in table[first].items()}
    selected = []
    for label in rows:
        index = actual.get(publication_token(label))
        row = {column:'NA' for column in columns}
        row[first] = label
        if index is None:
            auxiliary_issue(item['key'],label,'Required publication row unavailable.')
        else:
            for column in columns[1:]:
                if column in table:
                    row[column] = table.loc[index,column]
                else:
                    auxiliary_issue(item['key'],column,'Required publication column unavailable.')
        selected.append(row)
    extra_columns = [c for c in table if c not in columns]
    extra_rows = [str(value) for value in table[first] if publication_token(value) not in {publication_token(r) for r in rows}]
    if extra_columns or extra_rows:
        auxiliary_record(item['key'],'presentation_selection','',source,
                         detail=json.dumps({'omitted_columns':extra_columns,'omitted_rows':extra_rows}))
    return pd.DataFrame(selected,columns=columns)

def paper_auxiliary_model(names, source):
    """Resolve model 4 (or threshold model 5) from saved IDs, never average models."""
    role = source.get('paper_role','4')
    configured = MODEL_IDS_BY_DATASET.get(PAPER_DATASETS['random'],{}).get(role)
    if configured is not None:
        if configured not in names:
            raise ValueError(f'Configured model {configured!r} is absent from the auxiliary source.')
        return configured
    candidates = CUSTOM_MODEL_CANDIDATES if role=='4' else [TUNED_LIGHTGBM]
    matches = [name for name in candidates if name in names]
    if not matches and role=='4':
        matches = [str(name) for name in names if any(p in CUSTOM_MODEL_CANDIDATES for p in str(name).split('__'))
                   and not re.search(r'imagemol|uni[-_]?mol|gnn|gine|gat|gcn|mpnn|graphsage',str(name),re.I)]
    if len(matches)==1:
        return matches[0]
    if len(matches)>1:
        try:
            chosen = resolve_role(PAPER_DATASETS['random'],role)
            if chosen in matches:
                return chosen
        except ValueError:
            pass
    raise ValueError(f'Set model to the exact model {role} ID in this auxiliary source; saved IDs={list(names)}.')

def canonical_fold_columns(frame, source):
    """Normalize saved training/evaluation schemas, including combined exports."""
    groups = {
        'fold': ['outer_fold', 'outer_fold_id', 'fold_id'],
        'model': ['model_id'],
        'subset': ['evaluation_subset', 'task'],
        'parameter': ['param', 'parameter_name'],
    }
    rename = {}
    for target, aliases in groups.items():
        if target in frame or source.get(target+'_column', target) != target:
            continue
        present = [name for name in aliases if name in frame]
        if present:
            # evaluation_subset explicitly identifies the population; task may only
            # say classification/regression and therefore must not take precedence.
            rename[present[0]] = target
    return frame.rename(columns=rename)

def expanded_parameter_frame(frame, source):
    candidates = [c for c in ['parameters_json','best_parameters_json','best_params',
                              'best_params_json','best_parameters','params','params_json'] if c in frame]
    configured = source.get('parameters_json_column')
    if configured:
        if configured not in frame:
            raise ValueError(f'Parameter dictionary column is absent: {configured}')
        candidates = [configured]
    def parse(value):
        if isinstance(value, dict):
            return value
        if pd.isna(value) or not str(value).strip():
            return None
        try:
            result = json.loads(value)
        except (TypeError,ValueError):
            result = ast.literal_eval(value)
        if not isinstance(result,dict):
            raise ValueError('Saved parameters must be a JSON/Python-literal dictionary.')
        return result
    parsed_rows = []
    for _, row in frame.iterrows():
        values = [parse(row[c]) for c in candidates]
        values = [value for value in values if value is not None]
        if len(values)>1 and any(value != values[0] for value in values[1:]):
            raise ValueError('Populated parameter dictionaries disagree; specify parameters_json_column.')
        parsed_rows.append(values[0] if values else {})
    if candidates:
        expanded = pd.json_normalize(parsed_rows).set_index(frame.index)
        for name in expanded:
            destination = 'best_params.'+name
            if destination in frame:
                a,b = pd.to_numeric(frame[destination],errors='coerce'),pd.to_numeric(expanded[name],errors='coerce')
                if not np.allclose(a,b,equal_nan=True,atol=1e-12,rtol=1e-10):
                    raise ValueError(f'Repeated parameter values disagree: {name}')
            else:
                frame[destination] = expanded[name]
    return frame

def looks_like_presentation(table,item):
    layout = publication_layout(item)
    if layout is None:
        return True
    if any(name in table for name in ['model','model_id','metric','best_params','params_json']):
        return False
    columns, rows = layout
    names = {publication_token(name) for name in table}
    return all(publication_token(name) in names for name in columns)

def build_auxiliary_table(item):
    key = item['key']
    entry = AUXILIARY_TABLE_FILES.get(key)
    routed = False
    source_override = None
    if entry:
        specifier = dict(entry) if isinstance(entry,dict) else {'path':entry}
        path = auxiliary_path(specifier.get('path'))
        if not path.is_file():
            raise AuxiliaryUnavailable(f'Auxiliary source is missing: {path}')
        table = pd.read_csv(path,sep='\t' if path.suffix.lower()=='.tsv' else ',',dtype=str,keep_default_na=False) if path.suffix.lower() in ['.csv','.tsv'] else None
        if table is not None and (specifier.get('format')=='presentation' or looks_like_presentation(table,item)):
            if 'subset' in table and item['subset']:
                if set(table['subset'].map(subset_name)) != {item['subset']}:
                    raise ValueError(f'{key}: presentation subset does not match {item["subset"]}.')
            table = align_publication_table(table,item,path)
            auxiliary_record(key,'presentation_csv','',path)
            return table,f'Saved presentation values selected into the fixed publication layout from {path}.','Loaded saved table'
        if key not in ['supplementary_08','supplementary_09','supplementary_10','supplementary_11','supplementary_12']:
            raise ValueError(f'{key}: this is not a finished publication table. Configure the direct dataset/descriptor sources instead.')
        source_override = {k:v for k,v in specifier.items() if k!='format'}
        routed = True
    if key=='main_07':
        table,note = build_dataset_counts(key)
    elif key=='supplementary_07':
        table,note = build_descriptor_counts(key)
    elif key=='supplementary_08':
        table,note = build_search_spaces(key,source={**SEARCH_SPACE_SOURCE,**(source_override or {})})
    elif key=='supplementary_09':
        table,note = build_hyperparameters(key,source={**HYPERPARAMETER_SOURCE,**(source_override or {})})
    elif key in ['supplementary_10','supplementary_11']:
        table,note = build_fold_metrics(key,source={**PER_FOLD_SOURCES[key],**(source_override or {})})
    elif key=='supplementary_12':
        table,note = build_thresholds(key,source={**THRESHOLD_SOURCE,**(source_override or {})})
    else:
        raise AuxiliaryUnavailable('Supply a finished reference table or keep it excluded.')
    table = align_publication_table(table,item)
    if routed:
        note += ' The legacy AUXILIARY_TABLE_FILES path was read as raw results and passed through this table builder.'
        auxiliary_record(key,'legacy_raw_source','',source_override['path'],detail='Routed raw results through the fixed publication builder.')
    return table,note,'Generated from source data'


"""Resolve only files inside the configured source folder and explain every load."""
def result_pair_paths(source):
    directory = Path(source['directory'])
    if 'pairwise_files' in source:
        return [Path(path) for path in source['pairwise_files']]
    if source.get('format') in TASK_RESULT_FILES:
        return [directory/TASK_RESULT_FILES[source['format']][1]]
    defaults = {'09':'paired_two_sided_permutation_holm_by_pair_and_dataset.csv',
                '11':'04_pairwise_model_comparisons.csv','long':'pairwise.csv'}
    preferred = directory/defaults.get(source.get('format'),'pairwise.csv')
    if preferred.is_file():
        return [preferred]
    known = [directory/name for name in defaults.values() if (directory/name).is_file()]
    if len(known)>1:
        raise ValueError('Several comparison files exist; set pairwise_files to the intended saved analysis.')
    return known or sorted(directory.glob('*_combined.csv'))

def saved_pair_frame(source, paths):
    metadata_path=Path(source.get('metadata_file',Path(source['directory'])/'analysis_metadata.json'))
    metadata=json.loads(metadata_path.read_text()) if metadata_path.is_file() else {}
    confidence=source.get('confidence',metadata.get('confidence',.95))
    frames = []
    for path in paths:
        if not path.is_file():
            continue
        raw = pd.read_csv(path)
        if 'p_adjusted' not in raw:
            frame = parse_pairs(raw,source,path,confidence)
        else:
            required = {'model_a','model_b','subset','metric','p_adjusted'}
            if not required.issubset(raw):
                raise ValueError(f'{path}: missing normalized comparison columns {required-set(raw)}.')
            frame = raw[['model_a','model_b','subset','metric']].copy()
            frame['subset'] = frame.subset.map(subset_name)
            frame['metric'] = frame.metric.map(metric_name)
            for name in ['value_a','value_b','difference','ci_low','ci_high','n','p_adjusted','p_raw']:
                frame[name] = numbers(choose_column(raw,[name]))
            frame['confidence'] = choose_column(raw,['confidence'],confidence)
            frame['test_method'] = choose_column(raw,['test_method'],source.get('test_method','method not supplied'))
            frame['holm_family'] = choose_column(raw,['holm_family'],source.get('holm_family','correction family not supplied'))
            frame['dataset'],frame['source_file'] = source['dataset'],str(path)
            frame['source_record'],frame['record_type'] = np.arange(1,len(frame)+1),'comparison'
        frames.append(frame)
    return pd.concat(frames,ignore_index=True) if frames else pd.DataFrame()

def estimates_from_saved_pairs(source, pairs):
    """Reuse saved model values only. Differences and p-values are never estimates."""
    if pairs.empty:
        return pd.DataFrame()
    candidates = []
    for row in pairs.itertuples(index=False):
        for model,value in [(row.model_a,row.value_a),(row.model_b,row.value_b)]:
            if pd.notna(value):
                candidates.append(dict(model=model,subset=row.subset,metric=row.metric,value=value,
                                       n=row.n,source_file=row.source_file))
    if not candidates:
        return pd.DataFrame()
    data = pd.DataFrame(candidates)
    records = []
    for (model,subset,metric),group in data.groupby(['model','subset','metric'],sort=False):
        if not np.allclose(group.value,group.value.iloc[0],atol=1e-8,rtol=1e-6) or group.n.dropna().nunique()>1:
            raise ValueError(f'{source["dataset"]}/{subset}/{model}/{metric}: pair-specific values/counts differ. Supply the actual pooled estimates_file; do not mix paired cohorts.')
        row = group.iloc[0].to_dict()
        row.update(dataset=source['dataset'],format='saved_pair_values',record_type='estimate',
                   metric_model=model,configured_reference=source.get('reference',''),
                   source_file='; '.join(group.source_file.unique()),
                   estimate_provenance='Saved A/B metric values; agreement across pairs checked.',
                   distance_definition=source.get('distance_definition',DISTANCE_09))
        counts = group.n.dropna()
        row['n'] = counts.iloc[0] if len(counts) else np.nan
        records.append(row)
    return pd.DataFrame(records)

def read_configured_result_source(source):
    """Return data plus one diagnostic row, including optional/missing sources."""
    directory = Path(source['directory'])
    configured = source.get('format','09')
    metrics_names = {'09':'pooled_metrics_selected_models_long.csv','11':'02_model_metrics.csv','long':'estimates.csv'}
    metrics_names.update({fmt: files[0] for fmt, files in TASK_RESULT_FILES.items()})
    expected = Path(source['estimates_file']) if source.get('estimates_file') else directory/metrics_names.get(configured,metrics_names['09'])
    audit = dict(dataset=source['dataset'],configured_format=configured,directory=str(directory.resolve()),
                 expected_estimates=str(expected),estimates_file='',pairwise_files='',status='',
                 estimate_rows=0,comparison_rows=0,subsets='',detail='',available_csv_files='')
    try:
        if directory.is_dir():
            audit['available_csv_files'] = '; '.join(path.name for path in sorted(directory.glob('*.csv')))
        effective = dict(source)
        notes = []
        chosen = expected if expected.is_file() else None
        if chosen is None and not source.get('estimates_file') and configured not in TASK_RESULT_FILES:
            candidates = [(fmt,directory/name) for fmt,name in metrics_names.items() if fmt not in TASK_RESULT_FILES and (directory/name).is_file()]
            if len(candidates)>1:
                raise ValueError('Several possible estimate files exist; set estimates_file explicitly.')
            if candidates:
                effective['format'],chosen = candidates[0]
                notes.append(f'{source["dataset"]}: detected {chosen.name} as format {effective["format"]}.')
        pairs_paths = result_pair_paths(effective)
        # When point CSV layout changes, retain an explicitly selected comparison file,
        # or the comparison file belonging to the original configured reader when present.
        if 'pairwise_files' not in source and effective.get('format')!=configured:
            original_paths = result_pair_paths(source)
            if original_paths:
                pairs_paths = original_paths
        audit['pairwise_files'] = '; '.join(str(path) for path in pairs_paths)
        if chosen is not None:
            effective['estimates_file'] = str(chosen)
            if pairs_paths:
                effective['pairwise_files'] = [str(path) for path in pairs_paths]
            estimates,pairs,notices = load_result_source(effective)
            audit['estimates_file'] = str(chosen)
            audit['status'] = 'Loaded'
            notes.extend(notices)
        elif source.get('recover_pairwise_estimates',False) and pairs_paths and configured not in TASK_RESULT_FILES:
            pairs = saved_pair_frame(source,pairs_paths)
            estimates = estimates_from_saved_pairs(source,pairs)
            if estimates.empty:
                raise FileNotFoundError(f'{expected} is absent; comparison files do not contain saved A/B metric values.')
            audit['estimates_file'] = 'saved A/B values in '+audit['pairwise_files']
            audit['status'] = 'Loaded saved paired values'
            notes.append(f'{source["dataset"]}: reused saved A/B metric values from comparisons after checking agreement. No metric was calculated from p-values/differences.')
        else:
            extra = (' Comparison files were found, but a permutation folder alone does not supply pooled estimates.' if pairs_paths else '')
            raise FileNotFoundError(f'Expected estimates file not found: {expected}.{extra} Set estimates_file to the actual metrics CSV, or correct format/directory.')
        audit.update(estimate_rows=len(estimates),comparison_rows=len(pairs),
                     subsets='; '.join(sorted(estimates.subset.unique())),detail=' '.join(notes))
        return estimates,pairs,notes,audit
    except FileNotFoundError as error:
        audit.update(status='Missing required source' if source.get('required',True) else 'Missing optional source',detail=str(error))
        return pd.DataFrame(),pd.DataFrame(),[str(error)],audit
    except (ValueError,KeyError,pd.errors.ParserError) as error:
        audit.update(status='Invalid source',detail=str(error))
        return pd.DataFrame(),pd.DataFrame(),[str(error)],audit

def merge_source_records(frame, kind):
    """Combine explicitly configured exports; reject conflicting overlapping results."""
    if frame.empty:
        return frame
    frame = frame.copy()
    if kind == 'comparison':
        reverse = frame.model_a.astype(str) > frame.model_b.astype(str)
        for a, b in [('model_a', 'model_b'), ('value_a', 'value_b')]:
            frame.loc[reverse, [a, b]] = frame.loc[reverse, [b, a]].to_numpy()
        old_low = frame.loc[reverse, 'ci_low'].copy()
        frame.loc[reverse, 'ci_low'] = -frame.loc[reverse, 'ci_high']
        frame.loc[reverse, 'ci_high'] = -old_low
        frame.loc[reverse, 'difference'] *= -1
        keys = ['dataset', 'subset', 'model_a', 'model_b', 'metric']
        numeric = ['value_a', 'value_b', 'difference', 'ci_low', 'ci_high', 'n', 'p_adjusted', 'p_raw', 'confidence']
        text_fields = ['test_method', 'holm_family']
    else:
        keys = ['dataset', 'subset', 'model', 'metric']
        numeric = ['value', 'n', 'ci_low', 'ci_high', 'confidence']
        text_fields = ['configured_reference', 'publication_role']
    if not frame.duplicated(keys).any():
        return frame
    records = []
    for identity, group in frame.groupby(keys, sort=False, dropna=False):
        row = group.iloc[0].copy()
        for name in numeric:
            if name not in group:
                continue
            values = pd.to_numeric(group[name], errors='raise').dropna()
            tolerance = 0 if name == 'n' else 1e-10
            relative = 0 if name == 'n' else 1e-8
            if len(values) and not np.isclose(values, values.iloc[0], atol=tolerance, rtol=relative).all():
                raise ValueError(f'Conflicting {kind} exports for {identity}, field {name}. Select the intended analysis; sources: {group.source_file.tolist()}')
            row[name] = values.iloc[0] if len(values) else np.nan
        for name in text_fields:
            if name not in group:
                continue
            values = [str(value).strip() for value in group[name].dropna() if str(value).strip()]
            if len(set(values)) > 1:
                raise ValueError(f'Conflicting {name} for {identity}: {values}. Keep one correction family/reference.')
            if values:
                row[name] = values[0]
        row['source_file'] = '; '.join(dict.fromkeys(group.source_file.dropna().astype(str)))
        records.append(row)
    return pd.DataFrame(records).reset_index(drop=True)


def result_file_inventory(sources):
    """Inspect configured folders; identify candidate files without using them as results."""
    records = []
    for index, source in enumerate(sources, 1):
        folder = Path(source['directory'])
        audit = source_loading_report.iloc[index-1] if len(source_loading_report) >= index else {}
        selected = str(audit.get('estimates_file', ''))
        paths = sorted(folder.glob('*.csv'))
        if source.get('estimates_file'):
            explicit = Path(source['estimates_file'])
            if explicit.is_file() and explicit not in paths:
                paths.append(explicit)
        if not paths:
            records.append(dict(source_entry=index, dataset=source['dataset'], file=str(folder),
                                kind='Folder absent or no CSV files', models='', used_for_estimates=False, detail=''))
        for path in paths:
            try:
                columns = pd.read_csv(path, nrows=0).columns.tolist()
                names = set(columns)
                if {'model', 'metric', 'estimate', 'n_compounds', 'ci_low', 'ci_high'} <= names:
                    kind = 'Native Kp,uu estimates (kpuu)'
                elif {'model', 'family', 'subset', 'metric', 'estimate', 'n_subset', 'n_metric'} <= names:
                    kind = 'Native pooled hERG estimates (herg)'
                elif {'model', 'reference', 'metric', 'delta', 'p_holm', 'n_compounds'} <= names:
                    kind = 'Native Kp,uu paired statistics (kpuu)'
                elif {'reference', 'comparison', 'family', 'subset', 'metric', 'delta', 'p_holm'} <= names:
                    kind = 'Native hERG paired statistics (herg)'
                elif {'model', 'subset', 'metric', 'value'} <= names:
                    kind = 'Pooled long estimates (09/long)'
                elif {'model_a', 'model_b', 'metric'} <= names:
                    p_columns = {'p_adjusted', 'permutation_p_two_sided_holm_pair_dataset', 'permutation_p_two_sided_holm', 'p_bootstrap_holm'}
                    kind = 'Paired comparisons with adjusted-p column' if names & p_columns else 'Pairwise audit: no recognized adjusted-p column'
                elif names & {'predicted_logBB', 'predicted_bbb_class', 'score_nonpenetrant'}:
                    kind = 'Predictions: evaluate upstream before using as table statistics'
                elif 'model' in names and names & {'fold', 'outer_fold'}:
                    kind = 'Per-fold results: not pooled external estimates'
                elif 'model' in names and any(metric_name(c) in ALL_METRICS for c in columns):
                    kind = 'Wide estimates (11)'
                else:
                    kind = 'Other audit/metadata CSV'
                id_columns = [c for c in ['model', 'model_a', 'model_b', 'reference', 'comparison'] if c in names]
                model_ids = []
                if id_columns:
                    ids = pd.read_csv(path, usecols=id_columns, dtype=str)
                    model_ids = sorted(set(ids.stack().dropna()))
                used = selected == str(path) or selected == str(path.resolve())
                records.append(dict(source_entry=index, dataset=source['dataset'], file=str(path), kind=kind,
                                    models='; '.join(model_ids), used_for_estimates=used, detail=''))
            except (ValueError, OSError, pd.errors.ParserError, pd.errors.EmptyDataError) as error:
                records.append(dict(source_entry=index, dataset=source['dataset'], file=str(path), kind='Unreadable CSV',
                                    models='', used_for_estimates=False, detail=str(error)))
    return pd.DataFrame(records)


def publication_role_coverage():
    records = []
    for item in PAPER_TABLES:
        if item['style'] == 'auxiliary' or item['key'] in TABLES_TO_SKIP:
            continue
        group = estimates[(estimates.dataset == item['dataset']) & (estimates.subset == item['subset'])]
        for role in item['roles']:
            model = resolve_role(item['dataset'], role)
            present = set(group.loc[group.model.eq(model) & group.value.notna(), 'metric']) if model else set()
            missing_metrics = [metric for metric in item['metrics'] if metric not in present]
            records.append(dict(table=table_label(item['key']), dataset=item['dataset'], subset=item['subset'], role=role,
                                column=ROLE_LABELS[role], model=model or '',
                                status='Complete estimates' if not missing_metrics else ('Model ID not loaded/matched' if model is None else 'Missing estimates in this population'),
                                missing_metrics='; '.join(missing_metrics)))
    return pd.DataFrame(records)
```

## 4. Assemble tables using fixed publication definitions

Main Table 4 and SI S2/S3 use **Butina** results. Main Table 5 uses the original Inxight
evaluation. SI S18-S21 use the expanded dataset and its corresponding Inxight evaluation.
All performance tables have metrics down the left and the paper's models across the top.
Missing model columns remain NA; completely absent sources are recorded in the manifest.

Custom-MTL main-table comparisons select the better of models 2 and 3 from unrounded
estimates, with exact ties preferring model 2. The original saved correction families are
retained. Alternative-model comparisons use the explicit native-threshold ImageMol model
IDs when available; base aliases are fallbacks, never sources of another pair's p-value.

Main 7 and S7-S12 use the direct readers/calculations configured above. Finished CSVs remain
optional overrides. Excluded reference tables retain their numbers without missing-data flags.

```python
"""Notebook cell: fixed publication table definitions and display-only formatting."""
import re

missing = []
table_manifest = []
auxiliary_audit = []
PAPER_CLASS_METRICS = CLASS_METRICS[:-1]
PAPER_REG_METRICS = REG_METRICS[:-1]
KPUU_METRICS = ['spearman_r', 'pearson_r', 'kendall_tau', 'roc_auc', 'pr_auc_nonpenetrant',
                'balanced_accuracy', 'np_precision', 'np_recall', 'np_f1', 'specificity', 'mcc', 'cohens_kappa']
HERG_CLASS_METRICS = ['accuracy', 'balanced_accuracy', 'blocker_precision', 'blocker_recall',
                      'blocker_f1', 'nonblocker_recall', 'nonblocker_precision', 'nonblocker_f1',
                      'roc_auc', 'average_precision', 'pr_auc_trapezoid', 'mcc', 'cohens_kappa',
                      'misclassification_rate', 'wrong_side_mean_distance']
ROLE_CANDIDATES = {
    '1c': ['dummy_prior_classification'], '1r': ['dummy_mean_regression'],
    '3': ['lightgbm_mse_regression'], '4': CUSTOM_MODEL_CANDIDATES, '5': [TUNED_LIGHTGBM],
    'gnn_custom': ['gnn_custom_mtl', 'gine_custom_mtl'], 'gnn_mse': ['gnn_mse', 'gine_mse'],
    # These native-threshold model IDs match the comparisons in SI Tables S13-S15.
    # Base-name aliases are used only if the explicit native-threshold ID is absent.
    'image_custom': ['imagemol_resnet18_frozen_custom_mtl_native_threshold', 'imagemol_resnet18_frozen_custom_mtl'],
    'image_mse': ['imagemol_resnet18_frozen_regression_native_threshold', 'imagemol_resnet18_frozen_regression'],
    'image_bbbp': ['imagemol_bbbp', 'imagemol_bbbp_classifier', 'imagemol_native_bbbp', 'imagemol_bbbp_native', 'imagemol_bbbp_released'],
}
ROLE_LABELS = {
    '1c': '1c: Classification baseline', '1r': '1r: Regression baseline',
    '2': '2: Retrained logBB_pred', '3': '3: LightGBM (MSE objective)',
    '4': '4: This model', '5': '5: Threshold-tuned LightGBM (MSE objective)',
    'gnn_custom': 'GNN, custom loss', 'gnn_mse': 'GNN, MSE',
    'image_custom': 'ImageMol, custom loss', 'image_mse': 'ImageMol, MSE',
    'image_bbbp': 'ImageMol BBBP classifier',
}

def spec(key, title, dataset=None, subset=None, style='auxiliary', roles=None, metrics=None, **kwargs):
    return dict(key=key, title=title, dataset=dataset, subset=subset, style=style,
                roles=roles or [], metrics=metrics or [], **kwargs)

C = ['1c', '2', '3', '4', '5']
R = ['1r', '2', '3', '4']
A = ['4', 'gnn_custom', 'gnn_mse', 'image_custom', 'image_mse']
D = PAPER_DATASETS
PAPER_TABLES = [
    spec('main_01', 'Classification Performance on Molecules Only Labelled for Their Binary Classification', D['random'], 'classification_only', 'main', C, PAPER_CLASS_METRICS),
    spec('main_02', 'Classification Performance on Both Regression and Classification Labelled Molecules in the Dataset', D['random'], 'all_classification', 'main', C, PAPER_CLASS_METRICS),
    spec('main_03', 'Regression Performance on Molecules with Experimental logBB Labels', D['random'], 'measured_regression', 'main', R, PAPER_REG_METRICS),
    spec('main_04', 'Classification Performance on Compounds with Binary Labels under Butina Cluster-Based Nested Cross Validation', D['butina'], 'classification_only', 'main', C, PAPER_CLASS_METRICS),
    spec('main_05', 'Classification Performance on an External Inxight Test Dataset', D['external'], 'all_classification', 'main', C, PAPER_CLASS_METRICS),
    spec('main_06', 'Top Six Most Important Mordred Feature Clusters as Determined by Shapley Analysis'),
    spec('main_07', 'Breakdown of Compounds in the B3DB and the Constructed Drug-Like Subset. Values in Brackets Indicate the Number of Compounds in Each Class for which logBB Values Are Available'),
    spec('supplementary_01', 'Metrics used to evaluate logBB prediction methods or BBB-penetrance classification models'),
    spec('supplementary_02', 'Classification performance on all compounds under Butina cluster-based nested cross validation', D['butina'], 'all_classification', 'main', C, PAPER_CLASS_METRICS),
    spec('supplementary_03', 'Regression performance under Butina cluster-based nested cross validation', D['butina'], 'measured_regression', 'main', R, PAPER_REG_METRICS),
    spec('supplementary_04', 'Classification performance of the druglike B3DB-subset-trained models on the Kp,uu transfer task', D['kpuu'], 'all_classification', 'kpuu', ['4', '3', '5'], KPUU_METRICS),
    spec('supplementary_05', 'Classification performance of the custom loss on an alternative ADMET task involving hERG', D['herg'], 'all_classification', 'herg', ['3', '4'], HERG_CLASS_METRICS),
    spec('supplementary_06', 'Regression performance of the custom loss on an alternative ADMET task involving hERG', D['herg'], 'measured_regression', 'herg', ['3', '4'], ['mae', 'mse', 'mde', 'r2', 'pearson_r']),
    spec('supplementary_07', 'Number of available descriptors before and after filtering'),
    spec('supplementary_08', 'Hyperparameters tuned and the corresponding search spaces'),
    spec('supplementary_09', 'Chosen hyperparameters from each of the five outer folds'),
    spec('supplementary_10', 'Regression performance of each outer-fold model on its respective test dataset', D['random'], 'measured_regression'),
    spec('supplementary_11', 'Classification performance of each outer-fold model on all compounds in its respective test dataset', D['random'], 'all_classification'),
    spec('supplementary_12', 'Values of the BBB class threshold chosen by the threshold-tuned LightGBM model'),
    # S13 and S14 follow their captions, not the swapped numeric blocks in the attached SI.
    spec('supplementary_13', 'Classification performance of alternative model architectures on classification-only molecules in their corresponding outer folds', D['random'], 'classification_only', 'alternative', A, CLASS_METRICS),
    spec('supplementary_14', 'Classification performance of alternative model architectures on all molecules in their corresponding outer folds', D['random'], 'all_classification', 'alternative', A, CLASS_METRICS),
    spec('supplementary_15', 'Regression performance of alternative model architectures on molecules labelled for their logBB', D['random'], 'measured_regression', 'alternative', A, REG_METRICS),
    spec('supplementary_16', 'Classification performance of alternative model architectures on the external Inxight dataset', D['external'], 'all_classification', 'alternative', A+['image_bbbp'], PAPER_CLASS_METRICS),
    spec('supplementary_17', 'Molecular-property inclusion criteria for the original and expanded training datasets'),
    spec('supplementary_18', 'Classification performance on molecules only labelled for their binary classification in the expanded subset of the B3DB', D['expanded'], 'classification_only', 'main', C, CLASS_METRICS),
    spec('supplementary_19', 'Classification performance on all molecules in the expanded subset of the B3DB', D['expanded'], 'all_classification', 'main', C, CLASS_METRICS),
    spec('supplementary_20', 'Regression performance on molecules with experimental logBB labels in the beyond-rule-of-five subset of the B3DB', D['expanded'], 'measured_regression', 'main', R, PAPER_REG_METRICS),
    spec('supplementary_21', 'Classification performance on an external Inxight test dataset for models trained on the expanded subset of the B3DB', D['expanded_external'], 'all_classification', 'main', C, PAPER_CLASS_METRICS),
]

def table_label(key):
    return ('Table S' if key.startswith('supplementary_') else 'Table ') + str(int(key.rsplit('_', 1)[1]))

def number(value, integer=False, trim=False):
    if pd.isna(value) or not np.isfinite(value):
        return 'NA'
    if integer:
        if not np.isclose(value, round(value), atol=1e-8, rtol=0):
            raise ValueError(f'Noninteger count cannot be displayed as an integer: {value}')
        return str(int(round(value)))
    value = round(float(value), DECIMALS)
    text = f'{0.0 if value == 0 else value:.{DECIMALS}f}'
    return text.rstrip('0').rstrip('.') if trim else text

def p_text(value, comparator=None, brackets=True):
    if pd.isna(value):
        text = 'p=NA'
    else:
        if not np.isfinite(value) or not 0 <= value <= 1:
            raise ValueError(f'Invalid p-value {value}; metric differences are not p-values.')
        text = 'p<0.001' if value < 0.001 else f'p={number(value)}'
    if comparator is not None:
        text += f'; vs {comparator}'
    return f'[{text}]' if brackets else text

def report(dataset, subset, model, reference, metric, issue):
    missing.append(dict(dataset=dataset, subset=subset, model=model, reference=reference, metric=metric, issue=issue))

def find_pair(dataset, subset, a, b, metric):
    if pairwise.empty or a is None or b is None:
        return None
    match = pairwise[(pairwise.dataset == dataset) & (pairwise.subset == subset) & (pairwise.metric == metric) &
                     (((pairwise.model_a == a) & (pairwise.model_b == b)) | ((pairwise.model_a == b) & (pairwise.model_b == a)))]
    if match.empty:
        return None
    row = match.iloc[0].copy()
    if row.model_a != a:
        row['value_a'], row['value_b'] = row.value_b, row.value_a
        row['difference'] = -row.difference
        row['ci_low'], row['ci_high'] = -row.ci_high, -row.ci_low
    row['model_a'], row['model_b'] = a, b
    return row

def resolve_role(dataset, role):
    names = set(estimates.loc[estimates.dataset == dataset, 'model'])
    override = MODEL_IDS_BY_DATASET.get(dataset, {}).get(role)
    if override is not None:
        if override not in names:
            raise ValueError(f'{dataset}: configured model {override!r} for role {role} is absent.')
        return override
    # Native hERG IDs remain unchanged; roles come from the run's manifest/settings.
    if 'publication_role' in estimates:
        declared_roles = estimates.publication_role.astype(str).str.replace(r'\.0$', '', regex=True)
        declared = estimates.loc[(estimates.dataset == dataset) & declared_roles.eq(role), 'model'].unique()
        if len(declared) > 1:
            raise ValueError(f'{dataset}: multiple saved IDs declare role {role}: {list(declared)}. Set MODEL_IDS_BY_DATASET explicitly.')
        if len(declared) == 1:
            return declared[0]
    if role == '2':
        source_refs = estimates.loc[estimates.dataset == dataset, 'configured_reference'].dropna().unique() if 'configured_reference' in estimates else []
        source_refs = [x for x in source_refs if str(x).strip()]
        configured = REFERENCE_BY_DATASET.get(dataset, source_refs[0] if len(source_refs) == 1 else None)
        candidates = [configured] if configured else [m for m in REFERENCE_CANDIDATES if m in names]
        if configured is not None and configured not in names:
            raise ValueError(f'{dataset}: configured reference {configured!r} is absent.')
        if len(candidates) > 1:
            raise ValueError(f'{dataset}: set MODEL_IDS_BY_DATASET[{dataset!r}]["2"] to one exact model ID: {candidates}')
    else:
        candidates = [m for m in ROLE_CANDIDATES[role] if m in names]
        if role == '4' and not candidates:
            candidates = [m for m in sorted(names) if any(p in CUSTOM_MODEL_CANDIDATES for p in m.split('__'))
                          and not re.search(r'imagemol|uni[-_]?mol|gnn|gine|gat|gcn|mpnn|graphsage', m, re.I)]
            if len(candidates) > 1:
                raise ValueError(f'{dataset}: set an exact custom-MTL model ID in MODEL_IDS_BY_DATASET: {candidates}')
    return candidates[0] if candidates else None

def value_for(group, model, metric):
    match = group[(group.model == model) & (group.metric == metric)]
    return float(match.value.iloc[0]) if len(match) else np.nan

def better_baseline(group, metric, models):
    candidates = [(role, value_for(group, models.get(role), metric)) for role in ['2', '3']]
    candidates = [(r, v) for r, v in candidates if pd.notna(v) and np.isfinite(v)]
    if not candidates:
        return None
    return (min if metric in LOWER_IS_BETTER else max)(candidates, key=lambda item: item[1])[0]

def verify_pair_estimates(pair, group, dataset, subset, metric):
    for model, saved in [(pair.model_a, pair.value_a), (pair.model_b, pair.value_b)]:
        pooled = value_for(group, model, metric)
        if pd.notna(pooled) and pd.notna(saved) and not np.isclose(pooled, saved, atol=1e-8, rtol=1e-6):
            raise ValueError(f'{dataset}/{subset}/{model}/{metric}: pooled and pairwise estimates disagree.')

def metric_label(metric, item):
    if item['style'] == 'kpuu':
        return {'spearman_r': "Spearman's R", 'pearson_r': "Pearson's R", 'kendall_tau': 'Kendall Tau',
                'balanced_accuracy': 'Balanced accuracy', 'np_precision': 'Precision (low Kp,uu)',
                'np_recall': 'Recall (low Kp,uu)', 'np_f1': 'F1 (low Kp,uu)',
                'specificity': 'Specificity (high Kp,uu)', 'cohens_kappa': "Cohen's kappa"}.get(metric, LABELS[metric])
    if item['style'] == 'herg':
        return {'accuracy':'Accuracy', 'balanced_accuracy':'Balanced accuracy', 'blocker_precision':'Blocker precision',
                'blocker_recall':'Blocker recall (sensitivity)', 'blocker_f1':'Blocker F1',
                'nonblocker_recall':'Nonblocker recall (specificity)', 'nonblocker_precision':'Nonblocker precision (NPV)',
                'nonblocker_f1':'Nonblocker F1', 'average_precision':'Average precision',
                'pr_auc_trapezoid':'PR AUC (trapezoidal)', 'mcc':'Matthews correlation coefficient',
                'cohens_kappa':"Cohen's kappa", 'misclassification_rate':'Misclassification rate',
                'wrong_side_mean_distance':'Mean boundary distance among misclassified compounds (pIC50)',
                'mde':'MDE', 'pearson_r':"Pearson's R"}.get(metric, LABELS[metric])
    if metric == 'cohens_kappa' and item['key'].startswith('main_'):
        return 'κ'
    return LABELS[metric]

def column_label(role, item):
    if item['style'] == 'herg':
        return {'3':'Model 3: LightGBM, MSE', '4':'Model 4: LightGBM, MTL'}[role]
    if item['style'] == 'kpuu':
        return {'4':'Model 4: Custom loss', '3':'Model 3: LightGBM, MSE', '5':'Model 5: LightGBM, MSE, tuned threshold'}[role]
    if role == '4' and item['key'] in ['main_01', 'main_02', 'main_03']:
        return '4: This model (outer fold test sets)'
    return ROLE_LABELS[role]

def paper_cell(item, group, models, role, metric, used):
    dataset, subset = item['dataset'], item['subset']
    model = models[role]
    estimate = value_for(group, model, metric)
    text = number(estimate, integer=metric == 'np_wrong_side_count', trim=item['style'] == 'main')
    if pd.isna(estimate):
        report(dataset, subset, model or role, '', metric, 'Saved point estimate undefined or unavailable.')
    reference_role = None
    ci = False
    if item['style'] == 'main':
        reference_role = better_baseline(group, metric, models) if role == '4' else '4' if role == '5' else None
        ci = role == '4' and SHOW_DIFFERENCE_CI
    elif item['style'] == 'alternative':
        reference_role = '4' if role not in ['4', 'image_bbbp'] else None
    elif item['style'] == 'kpuu':
        reference_role = '4' if role != '4' else None
    if reference_role is None:
        if item['style'] == 'main' and role == '4':
            report(dataset, subset, model or role, '', metric, 'Neither baseline is available for comparator selection.')
        return text
    reference = models.get(reference_role)
    pair = find_pair(dataset, subset, model, reference, metric)
    if pair is not None:
        verify_pair_estimates(pair, group, dataset, subset, metric)
        used.append(pair)
    else:
        report(dataset, subset, model or role, reference or reference_role, metric, 'Exact model-pair/metric comparison not exported.')
    parts = [text]
    if ci:
        if pair is not None and pd.notna(pair.ci_low) and pd.notna(pair.ci_high):
            integer = metric == 'np_wrong_side_count' and all(np.isclose(x, round(x), atol=1e-8, rtol=0) for x in [pair.ci_low, pair.ci_high])
            parts.append(f'({number(pair.ci_low, integer)}, {number(pair.ci_high, integer)})')
        else:
            parts.append('(CI unavailable)')
            report(dataset, subset, model or role, reference or reference_role, metric, 'Saved difference CI unavailable.')
    p = pair.p_adjusted if pair is not None else np.nan
    parts.append(p_text(p, reference_role if item['style'] == 'main' else None))
    if pd.isna(p):
        report(dataset, subset, model or role, reference or reference_role, metric, 'Saved adjusted p-value unavailable.')
    return (' ' if item['style'] == 'main' else '\n').join(parts)

def source_count(group, models):
    selected = group[group.model.isin([m for m in models.values() if m is not None])]
    counts = selected.n.dropna().unique() if 'n' in selected else []
    return int(counts[0]) if len(counts) == 1 else None

def performance_note(item, used, group, models):
    note = 'Saved performance estimates and adjusted p-values are displayed without recomputing statistics. '
    if item['style'] == 'main':
        note += ('Model 4 is compared separately for each metric with the better of models 2 and 3, '
                 'selected using unrounded saved estimates; exact ties prefer model 2. '
                 'Parentheses are confidence intervals for model 4 minus its stated comparator. '
                 'Model 5 p-values compare model 5 with model 4. ')
    elif item['style'] in ['alternative', 'kpuu']:
        note += 'Bracketed p-values compare the column model with model 4. '
        if item['style'] == 'alternative':
            note += 'The released ImageMol BBBP classifier has point estimates only. '
    else:
        note += 'The separate p-value column compares model 4 with model 3. '
    if used:
        methods = sorted({str(r.test_method) for r in used if pd.notna(r.test_method)})
        families = sorted({str(r.holm_family) for r in used if pd.notna(r.holm_family)})
        note += 'Saved test: ' + '; '.join(methods) + '. Original correction family: ' + '; '.join(families) + '. '
    note += ('Blockers are the positive class. ' if item['style'] == 'herg' else
             'Low-Kp,uu compounds are the positive class. ' if item['style'] == 'kpuu' else
             'Nonpenetrants are the positive class. ')
    if 'wrong_side_mean_distance' in item['metrics']:
        note += 'Distance values retain the upstream evaluator\'s saved population and boundary definition. '
    if item['style'] == 'kpuu' and 'format' in group and group['format'].eq('kpuu').any():
        note += ('S4 PR AUC is the saved trapezoidal area (pr_auc_trapezoidal_low_Kpuu); '
                 'average precision is a distinct saved metric and is retained in the statistics export. ')
    if item['style'] == 'herg' and 'format' in group and group['format'].eq('herg').any():
        note += ('S5 uses classification on all compounds; S6 uses regression on measured IC50 compounds. '
                 'N is the saved subset size; class-specific n_metric and the other subsets remain in the statistics export. ')
    if 'source_note' in group:
        selected_notes = group.loc[group.model.isin([m for m in models.values() if m is not None]), 'source_note']
        saved_notes = list(dict.fromkeys(str(value).strip() for value in selected_notes.dropna() if str(value).strip()))
        if saved_notes:
            note += ' '.join(saved_notes) + ' '
    counts = {ROLE_LABELS.get(role, role): sorted(set(group.loc[group.model == model, 'n'].dropna().astype(int)))
              for role, model in models.items() if model is not None}
    if len({tuple(v) for v in counts.values() if v}) > 1:
        note += 'Model-specific saved sample counts: ' + '; '.join(f'{k}: {v}' for k, v in counts.items()) + '. '
    if item['key'] in ['supplementary_13', 'supplementary_14']:
        note += 'The subset follows this table\'s caption; the two numeric blocks were swapped in the supplied SI. '
    note += 'NA / unavailable identifies a missing or undefined saved result; values are never copied from the paper.'
    return note

def build_paper_tables():
    if list(EXPECTED_FOLDS) != [1,2,3,4,5]:
        raise ValueError('The attached SI defines exactly five outer folds (1-5).')
    unknown_skips = set(TABLES_TO_SKIP) - {item['key'] for item in PAPER_TABLES}
    if unknown_skips:
        raise ValueError(f'Unknown TABLES_TO_SKIP keys: {sorted(unknown_skips)}')
    tables, captions, notes = {}, {}, {}
    for item in PAPER_TABLES:
        key, dataset, subset = item['key'], item['dataset'], item['subset']
        status = dict(table=table_label(key), key=key, dataset=dataset or '', subset=subset or '',
                      status='', detail='', source_models='')
        if key in TABLES_TO_SKIP:
            status.update(status='Excluded from regeneration', detail='Reference/manual table excluded by TABLES_TO_SKIP; number retained.')
            table_manifest.append(status)
            continue
        if item['style'] == 'auxiliary':
            missing_before = len(missing)
            try:
                table, note, origin = build_auxiliary_table(item)
            except AuxiliaryUnavailable as error:
                status.update(status='Source not supplied', detail=str(error))
                table_manifest.append(status)
                report(dataset or '', subset or '', key, '', '', str(error))
                continue
            except (ValueError,KeyError,TypeError) as error:
                status.update(status='Input needs attention',detail=str(error))
                table_manifest.append(status)
                report(dataset or '',subset or '',key,'','',str(error))
                continue
            tables[key], captions[key] = table, table_label(key) + '. ' + item['title'] + '.' 
            notes[key] = note
            files = sorted({record['source_file'] for record in auxiliary_audit
                            if record['table'] == key and record['source_file']})
            status.update(status='Partial: missing values or inputs' if len(missing)>missing_before else origin,
                          detail='; '.join(files))
            table_manifest.append(status)
            continue
        group = estimates[(estimates.dataset == dataset) & (estimates.subset == subset)]
        if group.empty:
            status.update(status='Source not supplied', detail='No saved estimates for this dataset/subset. See source_loading_report.csv for the configured source status.')
            table_manifest.append(status)
            report(dataset, subset, key, '', '', status['detail'])
            continue
        models = {role: resolve_role(dataset, role) for role in item['roles']}
        missing_before = len(missing)
        records, used = [], []
        for metric in item['metrics']:
            row = {'Metric' if item['style'] in ['herg', 'kpuu'] else 'Model': metric_label(metric, item)}
            for role in item['roles']:
                row[column_label(role, item)] = paper_cell(item, group, models, role, metric, used)
            if item['style'] == 'herg':
                pair = find_pair(dataset, subset, models.get('4'), models.get('3'), metric)
                if pair is not None:
                    verify_pair_estimates(pair, group, dataset, subset, metric)
                    used.append(pair)
                p = pair.p_adjusted if pair is not None else np.nan
                row['Adjusted p-value'] = p_text(p, brackets=False)
                if pd.isna(p):
                    report(dataset, subset, models.get('4') or '4', models.get('3') or '3', metric, 'Saved adjusted p-value unavailable.')
            records.append(row)
        tables[key] = align_publication_table(pd.DataFrame(records),item)
        n = source_count(group, models)
        count = ('N = ' if key.startswith('main_') else 'n=') + str(n) if n is not None else 'n not supplied or model-specific'
        captions[key] = table_label(key) + '. ' + item['title'] + f' ({count}).'
        notes[key] = performance_note(item, used, group, models)
        status.update(status='Partial: missing values or comparisons' if len(missing) > missing_before else 'Assembled from saved results', detail=count,
                      source_models=json.dumps(models, ensure_ascii=False))
        table_manifest.append(status)
    return tables, captions, notes

tables, table_captions, table_notes = build_paper_tables()
table_manifest = pd.DataFrame(table_manifest)
with pd.option_context('display.max_colwidth',120):
    display(table_manifest[['table','dataset','subset','status','detail']])
print(f'{len(tables)} tables assembled. Missing sources keep their publication numbers reserved; see table_manifest.csv.')
```

contains stacked tables; it is not one rectangular analysis dataset. Missing numbers and

`auxiliary_table_audit.csv` retains unrounded direct counts/fold values and their source paths.

```python
STYLE = """<style>
body {margin:24px; color:#111; background:#fff; font-family:'Times New Roman',Times,serif;}
.bbb-wrap {overflow-x:auto; margin:16px 0 30px;}
.bbb-table {border-collapse:collapse; font-size:12pt; width:100%; min-width:1000px; margin:6px 0;}
.bbb-table th,.bbb-table td {padding:6px 8px; vertical-align:top; text-align:left;}
.bbb-table thead th {background:#e5e5e5; font-weight:bold; border-top:1px solid #111; border-bottom:1px solid #111;}
.bbb-table tbody tr:last-child td {border-bottom:1px solid #111;}
.bbb-table th:first-child,.bbb-table td:first-child {min-width:120px;}
.bbb-table caption {caption-side:top; text-align:left; font-size:13pt; font-weight:bold; padding:6px 0;}
.bbb-table.si th,.bbb-table.si td {border:1px solid #333;}
.bbb-note {font-size:10pt; line-height:1.3; margin-top:8px;}
@media print {body {margin:0;} .bbb-wrap {overflow:visible; page-break-inside:avoid;}
.bbb-table {min-width:0; font-size:9pt;} .bbb-table th,.bbb-table td {padding:4px;}}
</style>"""

def render_table(table, caption, note, key):
    heading = ''.join(f'<th>{html.escape(str(c))}</th>' for c in table.columns)
    if key=='supplementary_09':
        heading = '<tr><th rowspan="2">Hyperparameter</th><th colspan="5">Fold</th></tr><tr>'+''.join(f'<th>{fold}</th>' for fold in range(1,6))+'</tr>'
    else:
        heading = '<tr>'+heading+'</tr>'
    body = ''.join('<tr>' +''.join(f'<td>{html.escape(str(value)).replace(chr(10), "<br>")}</td>'
                  for value in row)+'</tr>' for row in table.itertuples(index=False, name=None))
    css_class = 'bbb-table si' if key.startswith('supplementary_') else 'bbb-table'
    return (f"<div class='bbb-wrap'><table class='{css_class}'><caption>{html.escape(caption)}</caption>"
            f'<thead>{heading}</thead><tbody>{body}</tbody></table>'
            f"<p class='bbb-note'>{html.escape(note)}</p></div>")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
table_manifest.to_csv(OUTPUT_DIR/'table_manifest.csv', index=False)
pd.DataFrame(auxiliary_audit, columns=['table', 'field', 'fold', 'value', 'source_file', 'detail']).to_csv(
    OUTPUT_DIR/'auxiliary_table_audit.csv', index=False)
# Remove this notebook's stale individual tables after an input is removed/excluded.
for previous in (OUTPUT_DIR/'individual_tables').glob('*.csv'):
    if previous.stem in {item['key'] for item in PAPER_TABLES} and previous.stem not in tables:
        previous.unlink()
width = max([len(table.columns) for table in tables.values()] or [1])
with (OUTPUT_DIR/'all_manuscript_tables.csv').open('w', newline='', encoding='utf-8-sig') as handle:
    writer = csv.writer(handle)
    for key, table in tables.items():
        for row in [[table_captions[key]], list(table.columns), *table.itertuples(index=False, name=None), [table_notes[key]], []]:
            writer.writerow([*row, *(['']*(width-len(row)))])
individual = OUTPUT_DIR/'individual_tables'
individual.mkdir(exist_ok=True)
for key, table in tables.items():
    table.to_csv(individual/f'{key}.csv', index=False, encoding='utf-8-sig')
blocks = [render_table(table, table_captions[key], table_notes[key], key) for key, table in tables.items()]
document = ('<!doctype html><html><head><meta charset="utf-8"><title>Manuscript and SI tables</title>'
            +STYLE+'</head><body>'+''.join(blocks)+'</body></html>')
(OUTPUT_DIR/'all_manuscript_tables.html').write_text(document, encoding='utf-8')
missing_statistics = pd.DataFrame(missing, columns=['dataset', 'subset', 'model', 'reference', 'metric', 'issue']).drop_duplicates()
missing_statistics.to_csv(OUTPUT_DIR/'missing_statistics.csv', index=False)
for block in blocks:
    display(HTML(STYLE+block))
print(f'Saved {len(tables)} tables with publication numbering in {OUTPUT_DIR.resolve()}. Performance tests and confidence intervals were not recomputed.')
print('All loaded estimates, models, and comparisons remain in all_table_statistics.csv.')
print('Source gaps and reserved table numbers are listed in table_manifest.csv; missing statistics are listed separately.')
print('Direct-count inputs and unrounded auxiliary values are recorded in auxiliary_table_audit.csv.')
```
